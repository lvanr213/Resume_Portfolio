### <font color='blue'>Multi-dimensional (aka multivariate) functions</font> 

A multi-variate (or multi-dimensional) function is a function in more than variable: $f(x, y,\ldots)$. 

```python
from astropy.cosmology import LambdaCDM
import astropy.units as u

def d_l_astropy(z, H0, Om0, OmL):
    cosmo = LambdaCDM(H0=H0, Om0=Om0, Ode0=OmL)
    return cosmo.luminosity_distance(z=z) / u.Mpc
```

### <font color='blue'> $d_L(z, \Omega_{\rm m0},\Omega_\Lambda)$ as a 3-D function</font>  
    
    
Let's look at how $d_L$ depends on $z$, $\Omega_{\rm m0}$ and $\Omega_\Lambda$ (for a fixed $H_0$). For this, we will "tabulate" the function on a 3D grid of $z$, $\Omega_{\rm m0}$ and $\Omega_\Lambda$ values, while keeping $H_0$ fixed to specific values. 

### <font color='blue'>Using RBF to approximate 3-dimensional $d_L(z, \Omega_{\rm m0}, \Omega_\Lambda)$</font>

We can use RBF to approximate the function as a function of all 3 parameters: $z, \Omega_{\rm m0}, \Omega_\Lambda$. 

*Note:* dependence of $d_L$ on $H_0$ is trivial - it is just proportional to $1/H_0$ so we can easily rescale $d_L$ computed for say $H_0=70$ to a different value: $d^\prime_L = d_L \times 70/H_{0,\rm new}$. So we don't need to construct approximation for the 4-dimensional $d_L(z, H_0, \Omega_{\rm m0}, \Omega_\Lambda)$. 


```python
# number of training points along 1d, must be complex for np.grid
ntr = 30j  
# grid of training values for x
xgrid = np.mgrid[0.01:2.:ntr, 0.:1.:ntr, 0.:1.:ntr]
xtrain = xgrid.reshape(3, -1).T
# sample is a 2d array of shape [nsamples, ndim]
dltrain = np.zeros((xtrain[:,0].size))
for i, x in enumerate(xtrain):
    dltrain[i] = d_l_astropy(x[0], 70., x[1], x[2])

# use training points to RBF  
tstart = time()
rbf = RBFInterpolator(xtrain, dltrain, neighbors=1000, epsilon=0.01)
print(f'RBF initialization time = {(time()-tstart):8.4f} sec')
```

```python
ntest = 1000
xtest= np.zeros((ntest,3))
xtest[:,0] = np.random.uniform(0.01, 2., size=ntest) # redshifts
xtest[:,1] = np.random.uniform(0., 1., size=ntest) # Om0
xtest[:,2] = np.random.uniform(0., 1., size=ntest) # OmL

dltest = np.zeros((ntest)) 

# Now cycle through Om0 and OmL values, compute d_L and fill the dlgrid array with values
tstart = time()
for i, omd in enumerate(xtest[:,1]):
    dltest[i] = d_l_astropy(xtest[i,0], 70., omd, xtest[i,2])

print(f'direct d_L execution time = {(time()-tstart):8.4f} sec')
```

```python
# compute interpolated values at test points 
tstart = time()
dl_rbf_test = rbf(xtest)
print(f'RBF test point execution time = {(time()-tstart):8.4f} sec')

# compute fractional error
frac_err = dl_rbf_test/dltest - 1.0

print(f'min. frac. error = {frac_err.min():.4g}, max frac. error = {frac_err.max():.4g}')
```

This shows that we can get only a few percent accuracy with RBF in approximating 3D $d_L$ with 30^3 training points, but the time it takes to compute approximation for test locations, is 40 times longer than to compute the integral for that number of points. So approximation is not useful, as it is less accurate than direct integration and is much more expensive, so cannot save us computing time. 

### <font color='blue'>An astronomical example: reconstruction of an image corrupted by cosmic rays</font>


Consider an image in which useful information in some random fraction of pixels is lost for some reason (in real observations this can happen when a charged particle, called cosmic ray, hits a pixel). We want to use good pixels to reconstruct the values in the damaged pixels. 

This is a toy example, but nevertheless is quite realistic. Real astronomical images are taken using CCDs (similar to devices used in digital cameras, such as in your phone, but larger and more sensitive and less noisy). Cosmic rays - charged particles that stream about constantly, can hit a pixel in a CCD and destroy any useful information accumulated in it. If exposed for a long time to a high flux of cosmic rays, CCD image will have many bad pixels. 

Cosmic rays hit pixels in image randomly, so their effect can be approximated by making a random subset of pixels unusable ("black"). 
We can attempt to use good pixels to repair (or reconstruct) values in the pixels affected by the cosmic rays. Namely, the photon counts in good pixels are used as training data to construct approximation of 2d function (image). The approximation can then be used to assign values of the approximation to all of the bad pixels. 

In the code below <a href="https://scikit-image.org/">scikit-image</a> package is used, which will also be used later in the course. So please install it, if it's not installed in the Python distribution you use (see <a href="https://scikit-image.org/docs/stable/user_guide/install.html">here</a> for installation instructions. If you use Anaconda distribution, use conda option to install). 

```python
from codes.plotting import show_image
```

```python
from skimage import io

# read in RGB image of M101 - "the Pinwheel galaxy"
image = io.imread('https://astro.uchicago.edu/~andrey/classes/a211/img/m101_nasa.jpg')
# image size
nxp, nyp = image.shape[0], image.shape[1]
print(f"image size {nxp:d} x {nyp:d} pixels; pixel data type is {image.dtype}")
```

```python
show_image(image, figsize=3)
```

```python
# diplaying only the green channel of the 3 R,G,B channels
show_image(image[:,:,2], figsize=3)
```

```python
# make a copy of one of the channels for experimentation
image_cr = np.copy(image[:,:,2])

# clip pixel values to the range of [0-254]
image_cr = np.clip(image_cr, 0, 254)
```

```python
# corrupting a fraction of pixels fcr by "cosmic rays"
fcr = 0.7 # extreme example of 70% of pixels corrupted by cosmic rays
xr = np.random.uniform(size=np.shape(image_cr))

cr_pixels = (xr <= fcr) # get filter array with True values for pixels affected by cosmic rays
image_cr[cr_pixels] = 255 # mark CR-hit pixels as black 
```

```python
show_image(image_cr, figsize=3)
```

We can see that the image does not ressemble the original. It looks hopelessly lost. 



```python
# find indices of good pixels
inds_good = np.argwhere(image_cr <= 254)
# find indices of bad pixels affected by cosmic rays
inds_cr = np.argwhere(image_cr == 255)
```

```python
# we will use good pixels as the training sample for RBF interpolation
train_vals = image_cr[inds_good[:,0],inds_good[:,1]]
# training pixel indices
xtrain, ytrain = inds_good[:,0], inds_good[:,1]
```

The calculation in the next cell will take a couple of minutes to complete... 

```python
# RBF interpolation

tstart = time()
# initialize RBF interpolation object
rbf = RBFInterpolator(inds_good, train_vals)
print("RBF initialization took %.1f sec to complete"%(time()-tstart))
```

```python
# interpolate using RBF
image_cr_fixed = np.copy(image_cr)

tstart = time()
# RBF interpolate values for CR-corrupted pixels 
image_cr_fixed[inds_cr[:,0], inds_cr[:,1]] = rbf(inds_cr)

print("RBF interpolation took %.1f sec to complete"%(time()-tstart))
```

Reconstructed image:

```python
show_image(image_cr_fixed, figsize=3)
```

Looks almost like the original image: 

```python
# original image
show_image(image[:,:,2], figsize=3);
```

Although some details are lost and reconstructed image looks smoother than the original one, it is fair to say that we got most of the visual information about this galaxy back. 

### <font color='blue'>This is amazing, but there are caveats:</font>
    
#### <font color='blue'>RBF interpolation is expensive</font> 

It takes $O(N^3_{\rm train})$ operations to compute coefficients for $N_{\rm train}$ training points and then $O(N_{\rm test})$ operations to compute approximate function at the test points. 
    
You can notice either by running above cells or by looking at the timing results, that it took ~5 minutes on my laptop to construct the RBF approximation, even though this image has only 270 x 400 pixels. 

RBF constructs approximation that can be evaluated at any $x$. 

However, in practice one often faces a more limited need to interpolate values from non-evenly distributed points to points on an evenly spaced grid. 

For example, in the CR-corrupted image above the distribution of "good" pixels is random, but we want to reconstruct image on a regular pixel grid. This problem is simpler and can be handled more efficiently either by the local (piecewise) linear or low-polynomial order interpolation. 

### <font color='blue'>2D linear interpolation from the function values at evenly spaced grid of points</font>
    
Suppose we have a grid of evenly spaced points covering a rectangular 2-dimensional $x-y$ domain, $x_i$ and $y_i$ where $i,j=0,\ldots, N-1$,  and function values tabulated at the grid nodes: $f_{ij} = f(x_i, y_j)$. 

A piecewise linear interpolation from the function values at grid points to a given point $(x,y)$ uses a 2D linear polynomial to approximate the function within the vicinity of that point:

$$f(x,y)\approx p_{1}(x,y) = a_{00} + a_{10}x + a_{01}y + a_{11} xy.$$


To find the 4 coefficients $a_{00}$, $a_{10}$, $a_{01}$, $a_{11}$ of the local approximating polynomial, we need 4 function values: these are taken to be the function values at the grid points closest to the point $(x,y)$. We can think of the grid of points as forming a grid of rectangular cells with side length equal to the spacing of grid points in each axis, with each grid point located in the center of each of such "cells". 

<center>
<img width=500 src="https://astro.uchicago.edu/~andrey/classes/a211/img/cic_schema.png"></img>

Rescale coordinates as follows $x\rightarrow \tilde{x}=(x-x_i)/h_x$, $\tilde{y}\rightarrow y=(y-y_j)/h_y$. The rescaled $\tilde{x}$ and $\tilde{y}$ now lie in $[0,1]$. 

A piecewise 2D linear interpolation polynomial to approximate the function around $(x,y)$ then is (see 11_approximation_multid notebook): 

$$p_1(x,y) = (1 -\tilde{x})(1-\tilde{y})\,f_{i,j} + (1-\tilde{y})\tilde{x}\,f_{i+1,j} + (1-\tilde{x})\tilde{y}\,f_{i,j+1} + \tilde{x}\tilde{y}\,f_{i+1,j+1}.$$ 

$$= w_{i,j}\,f_{i,j} + w_{i+1,j}\,f_{i+1,j} + w_{i,j+1}\,f_{i,j+1} + w_{i+1,j+1}\,f_{i+1,j+1}.$$ 


#### <font color='blue'>Geometric interpretation</font>

Training functions from surrounding grid cells are averaged with weights given by the corresponding shaded areas:
<center>
<img width=500 src="https://astro.uchicago.edu/~andrey/classes/a211/img/cic_schema.png"></img>

Such interpolation is used in simulations of $N$-body systems in the <a href="https://en.wikipedia.org/wiki/Particle_Mesh">Particle Mesh</a> (PM) codes and Particle-in-Cell (PIC) codes for simulations of particle dynamics in plasma. 

A related approach is to interpolate mass or charge of particles onto neighboring grid cells in such codes is called Cloud-in-Cell (CIC) method. The most commonly used interpolation here is also linear. 

### <font color='blue'>Linear interpolation from irregularly spaced points onto evenly spaced grid points</font>

If we face the opposite task of interpolating from randomly spaced training points to the points on evenly spaced grid, we can reuse the above result and carry out linear interpolation by assigning each of the nearest grid points the value of the function at the random point with the corresponding weight:

Each random points adds a contribution to each of the 4 nearest cells as follows
\begin{eqnarray}
f_{i,j} & = & f_{i,j} &+& f(x, y)\, w_{i,j}; \\ 
f_{i+1,j}  &=&  f_{i+1,j} &+& f(x, y)\, w_{i+1,j};\nonumber \\
f_{i,j+1} & = & f_{i,j+1} &+& f(x, y)\, w_{i,j+1};\\
f_{i+1,j+1} & =&  f_{i+1,j+1} &+& f(x, y)\, w_{i+1,j+1};\nonumber \\
\end{eqnarray}

Such interpolation is implemented in SciPy's `griddata` function. 

### <font color='blue'>Piecewise interpolation using SciPy's `RegularGridInterpolator` class and `griddata` function</font>

If we can grid training data on a uniform grid in 2d, 3d or any dimension, we can use SciPy's [RegularGridInterpolator](
https://docs.scipy.org/doc/scipy/reference/generated/scipy.interpolate.RegularGridInterpolator.html#scipy.interpolate.RegularGridInterpolator) class to construct approximation that will interpolate from the training grid values to a given points within the same space and the same range of values probed by the training points. Interpolation can be linear, cubic, or quintic (fifth order). 

Example below shows how this class could have been used to arrange for a fast evaluation of $d_L$ by constructing an approximation and then using to compute many distances for arrays of redshift, $\Omega_{\rm m0}$, $\Omega_L$. 

### <font color='blue'>Approximating $d_L(z, \Omega_{\rm m0},\Omega_\Lambda)$ as a 3-D function</font>  
    
    
Let's look at how $d_L$ depends on $z$, $\Omega_{\rm m0}$ and $\Omega_\Lambda$ (for a fixed $H_0$). For this, we will "tabulate" the function on a 3D grid of $z$, $\Omega_{\rm m0}$ and $\Omega_\Lambda$ values, while keeping $H_0$ fixed to specific values. 

```python
# construct training sample
ntr = 10  
# grids of training values for z, Om0, OmL
zgrid = np.linspace(0.01, 2., ntr)
om0grid = np.linspace(0., 1., ntr) 
omlgrid = np.linspace(0., 1., ntr) 

# sample is a 3d array of shape (ntr, ntr, ntr)
dltrain = np.zeros((ntr,ntr,ntr))
for i, om0 in enumerate(om0grid):
    for j, oml in enumerate(omlgrid):
        dltrain[:,i,j] = d_l_astropy(zgrid, 70., om0, oml)
```

```python
from scipy.interpolate import RegularGridInterpolator
tstart = time()
dl_interp = RegularGridInterpolator((zgrid, om0grid, omlgrid), dltrain, method='cubic')
print(f'Grid interpolator initialization time = {(time()-tstart):8.4g} sec')
```

```python
# construct test points 
ntest = 10000
ztest = np.random.uniform(0.01, 2., size=ntest) # redshifts
om0test = np.random.uniform(0., 1., size=ntest) # Om0
omltest = np.random.uniform(0., 1., size=ntest) # OmL

dltest_apy = np.zeros((ntest)) 
# iterate through Om0 and OmL values, 
# compute d_L and fill the dlgrid array with values
tstart = time()
for i, z in enumerate(ztest):
    dltest_apy[i] = d_l_astropy(z, 70., om0test[i], omltest[i])

print(f'direct d_L execution time = {(time()-tstart):8.4f} sec')
```

```python
# compute approximate d_L
tstart = time()
dtest_int = dl_interp((ztest, om0test, omltest))
print(f'evaluated test d_L with interpolation in {time()-tstart:.4f} sec')


ferr = dtest_int/dltest_apy - 1.
print(f'frac. error min = {ferr.min():.4g}, max = {ferr.max():.4g}')
```

Cubic interpolation is accurate but as expensive as direct integration. Using the linear approximation is about 400 times faster than direct integration.

### <font color='blue'>2d function approximation summary</font>

* Accurate methods developed for approximation of 1D functions work in 2D



* The number of coefficients and the cost of calculations increases as the square of the polynomial order or number of training points. 
    


* Polynomial approximation becomes numerically unstable (overflows/underflows) for larger order. 
    

* These factors make application of polynomial methods in >2D dimensions difficult generally, but such approximations are used for specific problems sometimes, especially with Chebyshev nodes. 



### <font color='blue'>2d function approximation summary</font>

 
* Piecewise linear and spline approximations are numerically stable and robust methods for 2D, 3D, and sometimes 3D approximation. 
    
* RBF can be used in any number of dimensions but tends to be computationally expensive for more than 1000 training points due to $O(N_{\rm train}^3$ operations to compute the coefficients. 
    
    
* Majority of applications of function approximation in >2D involve Machine Learning methods, which are often crude and approximate, but do scale well with increasing $D$.
