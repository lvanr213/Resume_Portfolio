## <font color='blue'>Drawing random numbers from a distribution: what does this mean?</font> 
   
The histogram of $N$ numbers drawn from a pdf should approach the form of the pdf as $N$ increases. 

Example: uniform distribution defined on the interval $[a, b]$: 

$$u(x) = \begin{cases} \frac{1}{b-a}, for x\in [a,b)\\
0,\ {\rm otherwise}.
\end{cases}$$

```python
def uniform_pdf(x, a, b):
    assert(a < b)
    x = np.array(x)
    if len(x)==1: 
        if a <= x < b:
            return 1.0 / (b-a)
        else:
            return 0.0
    else:
        isel = (a <= x) & (x < b)
        puni = np.zeros_like(x)
        puni[isel] = 1.0 / (b-a)
        return puni
```

## <font color='blue'>Sampling special case pdfs: drawing Gaussian random numbers and distributions symmetric around zero (optional material)</blue>

The Gaussian numbers can be drawn using the inverse transform, but this is computationally expensive due to a relatively complicated cdf of the Gaussian distribution (error function). Thus, a more efficient method is usually used: most commonly used is the <a href="https://en.wikipedia.org/wiki/Box%E2%80%93Muller_transform">Box-Mueller method.</a>. 

If $u_1$ and $u_2$ are uniformly distributed random numbers in the interval $[0,1)$, the following two numbers are drawn from the Gaussian pdf of zero mean and variance $\sigma^2=1$:


$$g_1=\sqrt{-2\ln u_1}\cos 2\pi u_2;\ \ \ g_2=\sqrt{-2\ln u_1}\sin 2\pi u_2 $$


If we want Gaussian numbers that sample Gaussian pdf with a given mean $\mu$ and standard deviation $\sigma$, we can rescale the above numbers as (where $g$ is either $g_1$ or $g_2$):

$$\tilde{g}=\mu + \sigma\,g$$


Let's check this empirically. 

The method is named after George Edward Pelham Box and Mervin Edgar Muller, who <a href="https://projecteuclid.org/euclid.aoms/1177706645">proposed it in 1958</a>, although the method was previously developed by Raymond E. A. C. Paley and Norbert Wiener in 1934.

## <font color='blue'>Drawing  random numbers from a Gaussian pdf with mean $\mu$ and dispersion $\sigma$</blue>

Using Box-Mueller method.

```python
def gaussian_pdf(x, mu=0., sigma=1.):
    """
    returns value of the Gaussian pdf with input mu, sigma for input x 

    Parameters:
        x:     a float or a 1d numpy array of floats, input x value(s)
        mu:    a float, mean of the Gaussian (default 0)
        sigma: a float, st. deviation of the Gaussian (default 1.0)
    
    Returns:
        g(x): a float or a 1d numpy array of floats, 
              value(s) of the Gaussian pdf for input x value(s)
    """
    assert(sigma !=0.)

    return 1./np.sqrt(2.*np.pi)/sigma * np.exp(-0.5*((x-mu)/sigma)**2)
```

```python
# generate two vectors of nr pseudo-random numbers uniformly distributed in [0,1] (default)
nr = 100_000
u1 = np.random.uniform(size=nr)
u2 = np.random.uniform(size=nr)

# transform uniformly distributed numbers in u1 and u2 into Gaussian pdf samples g1 and g2
mu, sigma = 0, 1.0
g1 = mu + sigma*np.sqrt(-2.*np.log(u1)) * np.cos(2.*np.pi*u2)
g2 = mu + sigma*np.sqrt(-2.*np.log(u1)) * np.sin(2.*np.pi*u2)
```

### <font color='blue'>The ITS  method: visual illustration</font>

The formal mathematical proof of the ITS method can be found in the optional reading below. 
However, it's useful to get an intuition into how and why this method works using visual illustrations. 

The illustration below shows visually what the proof shows with math: the sampling of $x$ is defined by the nonlinear transformation via inverse cdf that turns a given interval $\Delta y$ that is the same along the $y$ axis (uniformly distributed numbers will have equal probability to be drawn from each interval $[y,y+\Delta]$ as long as $\Delta y$ is the same) into $\Delta x$ that varies with $x$. 

So uniform density of $y$ samples will turn into non-uniform density of $x$ samples. This is because the number of samples in each $[y,y+\Delta]$ and corresponding $[x,x+\Delta x]$ is the same by construction, but $\Delta y$ is constant along $y$ axis, while $\Delta x$ varies and thus the density of samples will vary with $x$ even if its uniform in $y$. 

```python
from codes.plotting import its_demo, plot_pdf, its_demo_animation
```

```python
def pdf_exp(x, beta=1.):
    '''
    Exponential pdf defined on the interval x=[0, inf]
    '''
    return beta*np.exp(-beta*x)

def cdf_exp(x, beta=1.):
    '''
    cdf of the exponential pdf defined on the interval x=[0, inf]
    '''    
    return 1.-np.exp(-beta*x)

def cdf_exp_inv(x, beta=1):
    '''
    inverse cdf of the exponential pdf defined on the interval x=[0, inf]
    '''    
    return -beta*np.log(np.maximum(1-x,1e-300))
```

## <font color='blue'>Volume of the $d$-dimensional sphere volume</font> 
    
The exact formula for the volume of $d$-dimensional sphere of radius $r$: 

$$V_{\rm d}(R)=\frac{2R^d\pi^{d/2}}{\Gamma(d/2)\,d},$$

where $\Gamma(x)$ is the  gamma function (can be computed with [<tt>scipy.special.gamma</tt>](https://docs.scipy.org/doc/scipy/reference/generated/scipy.special.gamma.html) function).

$$\Gamma(z) = \int\limits_0^\infty t^{z-1}e^{-t}dt.$$

$\Gamma(1/2)=\sqrt{\pi};$ $\Gamma(1)=1$. 

```python
from scipy.special import gamma

def dsphere_volume_exact(r=1., dim=1):
    """
    Compute volume of dim-dimensional sphere of radius r
    using the exact formula
    """
    return 2. * r**d * np.pi**(0.5*dim) / (gamma(0.5*dim) * dim)
    
    
```

## <font color='blue'>Monte Carlo computation of the $d$-dimensional sphere volume</font> 
    
* Distribute $N$ points uniformly in the $d$-dimensional cube of side length $2r$ centered on the center of the sphere (assuming it is at the origin). 
The coordinates of each point are then uniformly distributed random numbers in the range $[-r,r]$. 
<p>

* Estimate the fraction of random points, $f_r$, for which the distance to the origin is $<r$.
<p>

* Monte Carlo estimate of the sphere volume is: $V_{d,\rm MC} = f_r V_{\rm cube} = f_r (2r)^d$. 
<p>
    
<center>
<img width=500 src="https://etc.usf.edu/clipart/43400/43443/4c6_43443_lg.gif"></img>

```python
def dsphere_volume_mc(r=1., d=1, nrnd=1):
    # generate uniformly distributed points in a cube 
    # reshape it into array of shape (nrnd, d)
    xd = np.random.uniform(low=-r, high=r, size=nrnd*d).reshape((nrnd, d))

    # compute the Eucledian distance from these points to the origin
    rd = np.linalg.norm(xd, axis=1)

    # compute and return fraction of the points that have rd<r 
    # using NumPy filtering operation 
    # and multiply by d-dim cube volume (2r)^2
    return np.size(rd[rd<r]) / np.size(rd) * (2.*r)**d
```

```python
def dsphere_volume_mc(r = 1., d=1, nrnd=1):
    """
    Parameters:
    -----------
        r: real
            radius of the sphere
        d: integer
            number of dimensions
        nrnd: integer
            number of Monte Carlo samples to use in the estimate

    Returns:
    --------
        float: estimate of the d-dimensional sphere volume
    """
    # generate uniformly distributed points in a cube 
    # reshape it into array of shape (nrnd, d)
    xd = np.random.uniform(low=-r, high=r, size=nrnd*d).reshape((nrnd, d))

    # compute the Eucledian distance from these points to the origin
    rd = np.linalg.norm(xd, axis=1)

    # compute and return fraction of the points that have rd<r 
    # using NumPy filtering operation 
    # and multiply by d-dim cube volume (2r)^2
    return np.size(rd[(rd<r)]) / np.size(rd) * (2.*r)**d
```

## <font color='blue'>Volume of 3-d sphere</font> 
   


```python
# a simple check for 3d sphere of r=1, for which we know volume is 4pi/3
r = 1.; d = 3

v_exact = 4. * r**3 * np.pi / 3.

for n in [1000]:
    vd = dsphere_volume_mc(r=r, d=d, nrnd=n)
    ferr = abs(vd/v_exact - 1.0)
    print(f'n = {n:9d};  V_estimate = {vd:.6f}; V_exact = {v_exact:.6f};  ferr = {ferr:.4e}')
    
```

```python
# a simple check for 3d sphere of r=1, for which we know volume is 4pi/3
r = 1.; d = 3

v_exact = 4. * r**3 * np.pi / 3.

nr = np.random.randint(1000, 1000_000, size=100)
ferr = np.zeros_like(nr, dtype='float')

for i, n in enumerate(nr):
    vd = dsphere_volume_mc(r=r, d=d, nrnd=n)
    ferr[i] = abs(vd/v_exact - 1.0)
```

```python
plt.figure(figsize=(3,3))
plt.yscale('log')
#plt.xscale('log')
plt.xlabel(r'$N$'); plt.ylabel('fractional error')
plt.scatter(nr, ferr, s=5, c='darkslateblue')
#nmodel = 10**np.linspace(3, 6, 50)
#ferr_model = 0.02 * (nmodel/1e3)**(-0.5) 
#plt.plot(nmodel, ferr_model, ls='--')
plt.show()
```

## <font color='blue'>Calculation with fixed $N$ but increasing number of dimensions $d$</font> 


```python
r = 2.
nrnd = 1_000_000
dmax = 20
vd = np.zeros((dmax))
vd_exact = np.zeros_like(vd)

for d in range(1,dmax+1):
    vd[d-1] = dsphere_volume_mc(r=r, d=d, nrnd=nrnd)
    vd_exact[d-1] = dsphere_volume_exact(r=r, dim=d)
    
```

```python
def plot_vd_error(x, vd, vde, xlabel='$x$', figsize=(7.5,3)):
    # plot volume of the d-sphere and fractional error of the MC estimate
    fig, ax = plt.subplots(1, 2, figsize=figsize)
    #ax[0].set_xlim(tlims); 
    #ax[0].set_ylim(1.e-3,200.)
    ax[0].set_xlabel(xlabel); ax[0].set_ylabel(r'$V_{\rm d}$')
    ax[0].scatter(x, vd, marker='o', s=20, 
                  color='slateblue', label='MC estimate')
    ax[0].plot(x, vde, c='orangered', label='exact')
    ax[0].grid(ls='dotted', c='lightgray')
    ax[0].legend(frameon=False, loc='best')
    
    ax[1].set_xscale('log')
    ax[1].set_yscale('log')
    ax[1].set_xlabel(xlabel); ax[1].set_ylabel(r'fractional error')
    ax[1].plot(x, np.abs(vd/vde-1.), c='orangered')
    ax[1].grid(ls='dotted', c='lightgray')
    plt.show()
```

```python
d = np.linspace(1, dmax, dmax)

plot_vd_error(d, vd, vd_exact, xlabel='$d$')
```

## <font color='blue'>Ratio of the $d$-sphere to $d$-cube volumes</font> 
   

Volume of $d$-dimensional sphere of radius $R$: 
$$V_{\rm d}(R)=\frac{2R^d\pi^{d/2}}{d\,\Gamma(d/2)},$$
Volume of $d$-dimensional cube is $V_{\rm d, cube}=(2R)^d=2^d R^d.$

So the ratio of the volumes is

$$\frac{V_{\rm d, sph}}{V_{\rm d, cube}}=\frac{2^{1-d}\pi^{d/2}}{d\,\Gamma(d/2)},$$

where $\Gamma(x)$ is the  gamma function (can be computed with [<tt>scipy.special.gamma</tt>](https://docs.scipy.org/doc/scipy/reference/generated/scipy.special.gamma.html) function):

```python
from scipy.special import gamma

d = 1
2**(1-d) * np.pi**(d/2) / d / gamma (d/2)

```

###  <font color='blue'>The Differential Evolution method</font>

Developed by R. Storn and K. Price in 1997 (<a href="https://link.springer.com/article/10.1023%2FA%3A1008202821328">Storn & Price 1997</a>), the <a href="https://en.wikipedia.org/wiki/Differential_evolution">differential evolution algorithm</a> is relatively simple to implement, works in arbitrary number of dimensions, does not require function derivatives, allows imposing bounds on the domain, and is quite efficient.

This method uses "walkers" or "population members" that are started at random locations and using random numbers to advance them in successive. Due ot the use of random numbers this method is in the category of the Monte Carlo methods.

 See [here](https://www.sciencedirect.com/science/article/pii/S111001682100613X) for a recent review paper on the details and applications of this algorithm.

### <font color='blue'>Differential evolution algorithm</font>

The case of the bivariate Rosenbrock pdf. 5 population members

<p>
<center>
<img width=600 src="https://astro.uchicago.edu/~andrey/classes/a211/img/minimize_de_npop5.gif"></img>

Animation above shows evolution of population members in the DE method for the case of 5 members initialized with locations uniformly distributed in $[-10,10]$ in both coordinates to minimize 2D negative log of the Rosenbrock function, as above. 

It shows that the method is immensely more effective in finding the minimum at $\mathbf{x}=(1,1)$ than the random walk method above. This is achieved by members exchanging information about their locations when making step proposal. 

For example, we can see 2 groups converging in two different locations along the narrow valley and the second group is quickly "teleported" to the minimum location when their step proposal used members close to the minimum. 

Moreover, because, the step proposal is defined by $s\,(\mathbf{x}_{{\rm now}, r_2}-\mathbf{x}_{{\rm now}, r_3})$ and so as members get closer to the minimum the step size decreases because it is proportional to the difference of member locations. 
The members thus converge onto the minimum and determine it more and more accurately. 

### <font color='blue'>Description of the method</font>

Start with $N_{\rm pop}$ members of a population with $\mathbf{x}$ values randomly distributed in the problem domain (which each argument in $\mathbf{x}$ being one of the coordinates of each member). 

Vector of coordinates for each member is indexed from $0$ to $N_{\rm pop}-1$ and collection of all vectors $\mathbf{X}_{\rm now}=(\mathbf{x}_{\rm now,0},\mathbf{x}_{\rm now,1},\ldots,\mathbf{x}_{{\rm now},N_{\rm pop}-1})$ forms the current population, where each $\mathbf{x}_{\rm now,i}$ is a $d$-dimensional vector, where $d$ is the number of parameters of the minimized function.   

The algorithm then proceeds through steps (called iterations or evolutionary or generational cycles) that lead to evolution of the population towards function minimum.

The reason the method is called "differential evolution" is because the initial population of points in $D$ dimensions can be thought of as evolving towards a minimum during algorithm iterations. The evolution is differential because it is driven by the scaled differences between member locations. 

In depth description of the differential algorithm, its variations, tests, and applications can be found in a book by R. Storn, K. Price and J. Lampinen "*Differential evolution : a practical approach to global optimization*" <a href="https://catalog.lib.uchicago.edu/vufind/Record/8875926">available online</a> through U.Chicago library. 

 Each iteration (evolution cycle) consists of the "*mutation*" and "*selection*" stages for each point vector $\mathbf{x}_i$ in $\mathbf{X}_{\rm now}$ (each member of the population) that leads to formation of the new population $\mathbf{X}_{\rm next}$:

* *Mutation stage for element $i$:*  select three random indices from $0$ to $N_{\rm pop}-1$, $r_1$, $r_2$, $r_3$ *different* from $i$. 
Perturb element with index $r_3$ by the scaled difference of vectors with indices $r_2$ and $r_3$: 
$\mathbf{x}^\prime_i=\mathbf{x}_{{\rm now}, r_3} + \lambda_r\,(\mathbf{x}_{{\rm now}, r_2}-\mathbf{x}_{{\rm now}, r_1})$, where $\lambda_r$ is a random number (a common choice is the uniformly distributed random number in the range $[s,1]$ where $s$ is a parameter). 


* *Selection stage:* if $f(\mathbf{x}^\prime_i)\leq f(\mathbf{x}_{{\rm now},i})$, then $\mathbf{x}_{{\rm next},i}=\mathbf{x}^\prime_i$, else $\mathbf{x}_{{\rm next},i}=\mathbf{x}_{{\rm now},i}$ (no mutation). 

Once all members of the population went through mutation and selection stages go back to the next iteration and $\mathbf{X}_{\rm next}$ becomes $\mathbf{X}_{\rm now}$ and the cycle repeats. 

### <font color='blue'>Mutation (or step proposal) in more detail</font>

The idea is to use information in locations of 3 other randomly chosen population members. As members move towards the minimum, the steps using other member locations will tend to be towards the minimum as well.  Note that location of the member for which the new location is constructed this way is not used in this mutation stage.

<p>

<center>

### <font color='blue'>Convergence criterion</font>

The steps are stopped when a convergence criterion is satisfied. Several criteria can be used (or a combination thereof). A reasonable one is to check that the maximum of the absolute difference between current and new location is smaller than a threshold:

$$\max \vert \mathbf{x}_{\rm next}-\mathbf{x}_{\rm now}\vert<\rm\, atol,$$ 

where $\rm atol$ is a threshold we specify. This ensures a certain absolute accuracy with which we determine  values of function arguments at which minimum is located. 



***Note:*** during some steps, especially for small $N_{\rm pop}$, all proposed steps may be rejected so that $\mathbf{x}_{\rm next}=\mathbf{x}_{\rm now}$ and $\vert \mathbf{x}_{\rm next}-\mathbf{x}_{\rm now}\vert = 0$. This satisfies the stopping criterion, but we should not stop the steps if this occurs. Thus, the criterion should be 


$$0<\max \vert \mathbf{x}_{\rm next}-\mathbf{x}_{\rm now}\vert<\rm\, atol,$$ 



Sometimes convergence criterion related to the function value is used. I.e. absolute or relative fractional accuracy is imposed on the function 

### <font color='blue'>Choosing $s$ scaling factor</font>


The $s$ factor (or $\lambda$ in the mutation stage illustrDifferent members stepping to the location, $s$ for each population is chosen to be a random number drawn from a distribution. Common choices: 


* Gaussian distribution with the mean at some $s_{\rm mean}$ and dispersion $\sigma_s$.<p> 


* Uniform distribution $U(a,b)$. 


The choice that seems to work well in practice is $U(s,1)$, where $s\approx 0.1-0.3$. 

### <font color='blue'>For stable results a sufficiently large number of population members should be used</font>

### <font color='blue'>Crossover stage</font>

Incorporates at least some information for the current member for which we perform mutation into the new proposed location for this member. In this case information from the locations of more than 3 members is used. The stage is inserted after the mutation stage. 


* "***crossover stage***": Using current member location $\mathbf{x}_{i,\rm now}$ and location from the mutation stage $\mathbf{x}^\prime_i$ form the proposed new location as a combination (crossover) of these two locations:  $\mathbf{x}^{\prime\prime}_i=[x^{\prime\prime}_{0i}, x^{\prime\prime}_{1i}, \ldots, x^{\prime\prime}_{(D-1)i}]$, by selecting $j$-th element of $\mathbf{x}^{\prime\prime}_i$ to be the $j$-th element of $\mathbf{x}^\prime_i$ with the *recombination probability* $r$ ($r\in[0,1]$) and $j$-th element of $\mathbf{x}_{i,\rm now}$ with probability $1-r$. 


So on average over many steps the fraction $r$ of coordinates in new proposed locations will come from the location constructed during mutation stage and $(1-r)$ coordinates will stay the same as in the current location of the member. 


The actual algorithm for the crossover stage looks like this: 


* "***crossover stage***": Using current member location $\mathbf{x}_{i,\rm now}$ and location from the mutation stage $\mathbf{x}^\prime_i$ form the trial vector $\mathbf{x}^{\prime\prime}_i=[x^{\prime\prime}_{0i}, x^{\prime\prime}_{1i}, \ldots, x^{\prime\prime}_{(D-1)i}]$, where 

\begin{equation}
x^{\prime\prime}_{ji} = 
\begin{cases}
x^{\prime}_{ji}, \ {\rm if\ }r_j\leq \mathrm{cr\ or\ } j= \mathrm{ir}_i,\\
x_{{\rm now},ji}, \ {\rm otherwise\ }
\end{cases}
\end{equation}

and $r_j$ is the random floating point number uniformly distributed in the interval $[0,1]$ that is generated for the index $j$
which is called *recombination probability*. 

$\mathrm{ir}_i$ is the random integer uniformly distributed in the range $[0, D-1]$ generated for index $i$, which ensures that $\mathbf{x}^{\prime\prime}_i$ gets at least one element from $\mathbf{x}^\prime_i$. The crossover parameter $\mathrm{cr}\in [0,1]$ is a constant set by user. 


Evolutionary cycles continue until some convergence criterion is met. For example, maximum difference between vectors in $\mathbf{X}_{\rm next}$ and $X_{\rm now}$ is smaller than the specified threshold. 

Differential evolution method is implemented in <a href="https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.differential_evolution.html#scipy.optimize.differential_evolution"><tt>scipy.optimize.differential_evolution</tt></a> function. 

Crossover stage makes the algorithm more stable - stable results can be obtained for considerably smaller number of population members. 

### <font color='blue'>Minimum of the Rosenbrock function</font>

is at $[1, 1, \dots, 1]$ (where length is equation to the number of function arguments)


```python
def rosenbrock(x):
    """The natural log of the Rosenbrock "banana" function
    x is a vector of points in 2 or more dimensional space
    In this function x is an array of shape (ndim) containing 
    values of ndim function arguments. 
    
    Returns: 
        float, value of the ndim-dimensional Rosenbrock function for input x
    """
    return sum(100.0*(x[1:]-x[:-1]**2.0)**2.0 + (1-x[:-1])**2.0)
```

```python
from time import time 
from scipy.optimize import differential_evolution
# the total number of pop members will be popsize*5
popsize = 6

bounds = np.array([(-15, 15), (-10, 10), (-9,9), 
                   (-8, 8), (-9.5, 9.5)])

s, cr = 0.2, 1.
tstart = time()
res = differential_evolution(rosenbrock, bounds=bounds, mutation=[s, 1.], recombination=cr, 
                             atol=1.e-8, tol=0., strategy='rand1bin',
                             popsize=popsize, polish=False)

print(f"completed in {(time() - tstart):.3g} sec")
print("found minimum at:", res.x)
print(f"in {res.nfev:d} function evaluations")
print(40*'-')
```

### <font color='blue'>Differential Evolution is an example of a *heuristic algorithm*</font>

A heuristic algorithm is an algorithm which has proven its efficacy in solving a practical problem and elements of which have been selected by experimentation or heuristic arguments and not based on rigorous mathematical analysis (e.g., it is better to use information from location of multiple members as the "walk" towards the minimum than not to use such information). 

The choices in a heuristic algorithm may thus not be optimal for a given problem, but it is used as long as it does its job sufficiently well. 

Mathematical analysis is often done after heuristic algorithm. 

So specific choices of the mutation stage step are a result of experimentation. In fact, in such heuristic algorithms different choices can be done, which work better for specific problems. So heuristic algorithms usually consist of a class of methods, not of a single specific algorithm.

###  <font color='blue'>Initial distribution of population members</font>

Mutation stage of the algorithm requires that population members are in different locations, otherwise the difference $\mathbf{x}_{{\rm now}, r_2}-\mathbf{x}_{{\rm now}, r_1}$ will be zero and location of $\mathbf{x}_{{\rm now}, r_3}$ will simply be duplicated to location of $\mathbf{x}^\prime_i$ in the mutation stage.

Thus, initial locations of population members must be different. There are different ways to place population members initially. Assuming that we are searching for a minimum within some rectangular domain defined by the minimum and maximum values along each coordinate axis: $\mathbf{x}_{\rm min}$ and $\mathbf{x}_{\rm max}$, we can initialize the members as 

$$\mathbf{x}_0 = \mathbf{x}_{\rm min} + (\mathbf{x}_{\rm max}-\mathbf{x}_{\rm min}) \times \mathrm{rand}(0,1),$$

where $\mathrm{rand}(0,1)$ is a random number uniformly distributed from 0 to 1 (can be produced with <tt>np.random.uniform</tt>, for example). 

This choice makes sense if we do not know the location of the minimum and we are spreading the population members as widely as possible. To make the distribution of members more uniform, certain placement strategies can be used instead of random distribution above, such as <a href="https://www.johndcook.com/blog/2009/03/16/quasi-random-sequences-in-art-and-integration/">quasi-random sequences</a>.

Another possibility is to initialize locations of population members around a given point $\mathbf{x}_{0c}$ by using random numbers drawn from a Gaussian distribution centered at $\mathbf{x}_{0c}$ with some non-zero dispersion around it. 

 This choice makes sense if we have a good idea where the minimum is roughly, so that we can choose $\mathbf{x}_{0c}$ to be near the suspected location.
