### <font color='blue'>Typical steps involved in obtaining computational solution to a problem</font>
</center>
    

1. Definition of the problem that needs to solved.

2. Mathematical equations that describe the solution.

3. Algorithm to compute *approximate* solution on a computer (numerically). 

4. Implementation of the algorithm in code. 

5. Testing the code using a case with a known solution or comparing to another well-tested code. 

6. Applying the code to the problem at hand. 



<center>

### <font color='blue'>Computational algorithms always result in an approximate solution that has certain error (difference from the true solution)</font>
</center>
 
This is because 

1. Mathematical equations describing the model we calculate involve mathematical idealizations, such as limit $\Delta x\rightarrow 0$, which is not possible to achieve in real calculations. 
<br>

2. Floating point operations in the actual computer computations have [limited precision](https://floating-point-gui.de/formats/fp/) that limit the precision of any calculation.

## <font color='blue'>Equations: </font>

Evolution of position specified by the coordinate vector $\vec{r}=[x,y,z]$ is given by 
$$\frac{d\vec{r}}{dt}= \vec{v}.$$

Evolution of velocity is governed by the second Newton's law for a body of mass $m$ experiencing a force $\vec{F}$:     
$$\frac{d\vec{v}}{dt}=\vec{F}=m\vec{a}.$$

Often gravitational field at the position specified by coordinate vector $\vec{r}=[x,y,z]$ is specified by the potential $\phi(\vec{r})$ defined such that acceleration is given by 

$$\vec{a} = -\nabla \phi = -\left[\frac{\partial\phi}{\partial x}, \frac{\partial\phi}{\partial y}, \frac{\partial\phi}{\partial z}\right].$$

<center>
    
## <font color='blue'>Equations for the $x$-component: </font>

Consider equations for the $x$ coordinate and $x$-component of velocity. Equations for $y$ and $z$ will be similar. 
$$\frac{dx}{dt}= v_x(t),$$
    
 
$$\frac{dv_x}{dt}= a_x(t) = -\frac{\partial\phi}{\partial x}.$$

We can separate differentials $dx$ and $dt$: $dx = v_x(t) dt$ and take integral from $t_0$ to $t$ of both sides: 

$$\int\limits_{t_0}^t dx = \int\limits_{t_0}^t v_x(t) dt.$$

The first integral contains just the full differential and is simply equal to $x(t) - x(t_0)$ so we can write
the equation for volution of the $x$ coordinate as

$$x(t) = x_0 + \int\limits_{t_0}^t v_x(t) dt.$$



Integral gives the area under the curve within integration limits. Thus, the change $\Delta x = x(t)-x_0$ is equal to the area under the line $v_{x}(t)$ in the $v_x-t$ plane between $t_0$ and $t$. 

Calculating trajectory of a particle moving in a gravitational field thus requires numerical evaluation of the integral above. 

<center>
    
## <font color='blue'>Check for the case when acceleration is constant: $a_x={\rm const}$ </font>
</center>


For example, if particle is moving with a constant acceleration that velocity increases as a linear function of $t$: $v_{x,0} + a_{x,0}(t-t_0)$ and the area under $v_{x}(t)$ between $t_0$ and $t$ is the area of the trapezoid with sides $v_{x,0}$ and $v_{x,0} + a_{x,0}(t-t_0)$ is: 

$$\frac{1}{2}\left[v_{x,0} + v_{x,0} + a_{x,0}(t-t_0)\right](t-t_0)= v_{x,0} + \frac{1}{2} a_{x,0}(t-t_0)^2,$$

which is the expression from basic mechanics that should be familiar to you. 

<center>
    
## <font color='blue'>Motion with the constant acceleration: $a_x={\rm const}$ </font>
</center>

If particle experiences a constant acceleration, starting at the position with $x_0$ with velocity $v_{x,0}$ and acceleration $a_{x,0}={\rm const}$ at time $t_0$ the $v_x$ component will change as: 
    
$$v_x(t) = v_{x,0} + a_{x,0}(t-t_0),$$
    
while $x$ coordinate will change as 
    
$$x(t) = x_0 + v_{x,0}(t-t_0) + \frac{1}{2} a_{x,0}(t-t_0)^2.$$

Do these equations remind you of something?

<center>
</img>
</center>

## <font color='blue'>These are the first couple of terms of the Taylor series expansions of $x(t)$ and $v_x(t)$ arount $t_0$:</font>

$$v_x(t) = v_{x,0} + \dot{v}_{x,0}(t-t_0),$$
$$x(t) = x_0 + \dot{x}_{0}(t-t_0) + \frac{1}{2} \ddot{x}_{0}(t-t_0)^2.$$

In the case of constant acceleration the above expansions are exact, but if acceleration is not constant, there are other terms in the expansions: 

$$v_x(t) = v_{x,0} + \dot{v}_{x,0}(t-t_0)+\frac{1}{2} \ddot{v}_{x,0}(t-t_0)^2 + \mathcal{O}([t-t_0]^3),$$
$$x(t) = x_0 + \dot{x}_{0}(t-t_0) + \frac{1}{2} \ddot{x}_{0}(t-t_0)^2 + \frac{1}{6}\dddot{x}_0(t-t_0)^3 + \mathcal{O}([t-t_0]^4),$$

where notation $\mathcal{O}([t-t_0]^3)$ is a shorthand for "terms proportional to $(t-t_0)^3$ and higher powers of $(t-t_0)$."

In other words, we can also write the second equation for $x(t)$ like this 

$$x(t) = x_0 + \dot{x}_{0}(t-t_0) + \frac{1}{2} \ddot{x}_{0}(t-t_0)^2 +  \mathcal{O}([t-t_0]^3),$$

#### <font color='blue'>Sums of the 1, 2, 3, 4 terms of the Taylor expansion of $e^x$ around $t=1$</font>

## <font color='blue'>Time-stepping scheme: this is our algorithm</font>

Taylor expansion is more accurate when $\Delta t = t-t_0$ is small. Thus, we can approximate trajectory of a particle moving in a gravitational field by making a sequence of "steps" with some sufficiently small $\Delta t$ starting at some initial time $t_0$ with initial conditions for $x_0$ and $v_0$. 

For example, 
$$x_1 = x_0 + v_{x,0}\Delta t_0$$
$$v_{x,1} = v_{x,0} + a_{x,0}\Delta t_0$$
and repeating this again and again until some final desired time $t$ is reached. 
$$\ldots$$
$$x_{n+1} = x_n + v_{x,n}\Delta t_n$$
$$v_{x,n+1} = v_{x,n} + a_{x,n}\Delta t_n.$$

As a result, we get a sequence of values $x_i$, $v_{x,i}$ at discrete times $t_i$, which we can analyze (plot, etc.). 


### <font color='blue'>Implementation of a time-stepping sequence</font>

```python
def time_stepping(x0, v0, t0, tend, dt, *args, acc_func=None):

    # initialize time and arrays to keep coordinate and velocity components
    t = t0
    x, v = np.array(x0), np.array(v0)

    # Initialize lists to store times, coordimates, and velocities along trajectory
    time, xt, vt = [np.copy(t)], [np.copy(x0)], [np.copy(v0)]

    # Time stepping loop
    while t < tend:
        # Calculate acceleration components using the provided function
        a = np.array(acc_func(x, *args))

        # Update velocity and position
        x += v * dt
        v += a * dt

        # Update time
        t += dt

        # Append current state to the trajectory
        time.append(t)
        xt.append(np.copy(x))
        vt.append(np.copy(v))
        
    return np.array(time), np.array(xt), np.array(vt)

```

```python
def time_stepping(x0, v0, t0, tend, dt, *args, acc_func=None):
    """
    Implements time stepping for a particle's motion in three dimensions.

    This function updates the position and velocity of a particle over a series of time steps from t0 to tend.
    The acceleration of the particle at each time step is obtained by calling the acc_func function.

    Parameters:
    x0 (array-like): Initial coordinates of the particle at time t0, format [x, y, z].
    v0 (array-like): Initial velocity components of the particle at time t0, format [vx, vy, vz].
    t0 (float): Initial time.
    tend (float): End time.
    dt (float): Time step for the update.
    args: tuple of parameters to be passed to the acc_func 
    acc_func (function): A function that takes the current position (x) and velocity (v)
                         and returns the acceleration components as an array [ax, ay, az].

    Returns:
    array: arrays of time [shape =(nsteps)], positions [shape=(nsteps,3)], velocities [shape=(nsteps,3)].
    """

    # initialize time and arrays to keep coordinate and velocity components
    t = t0
    x, v = np.array(x0), np.array(v0)

    # Initialize lists to store times, coordimates, and velocities along trajectory
    time, xt, vt = [np.copy(t)], [np.copy(x0)], [np.copy(v0)]

    # Time stepping loop
    while t < tend:
        # Calculate acceleration components using the provided function
        a = np.array(acc_func(x, *args))

        # Update velocity and position
        x += v * dt
        v += a * dt

        # Update time
        t += dt

        # Append current state to the trajectory
        time.append(t)
        xt.append(np.copy(x))
        vt.append(np.copy(v))
        
    return np.array(time), np.array(xt), np.array(vt)
```

### <font color='blue'>Test case: acceleration from a point mass</font>

the potential of a point-mass of mass $M$ in the coordinate system where the mass is at the origin is 
$$\phi = -\frac{GM}{r},\ \ \ {\rm where}\ \ \ r=\sqrt{x^2 +y^2 + z^2}.$$
The $x$-component of the acceleration a particle experiences at the point $[x, y, z]$ is thus
$$
\begin{aligned}
a_x =& -\frac{\partial\phi}{\partial x} = \frac{\partial}{\partial x}\left(\frac{GM}{r}\right)\\ =&-\frac{GM}{r^2}\frac{\partial r}{\partial x}=-\frac{GM}{r^3}\, x
\end{aligned}$$

because
$$\frac{\partial r}{\partial x} = \frac{1}{2}\,\frac{1}{r}\frac{\partial (x^2 + y^2+z^2)}{\partial x}=\frac{1}{2}\,\frac{1}{r}\, 2x=\frac{x}{r}.$$

The $y$- and $z$-components are calculated similarly so that
$$\vec{a} = [a_x, a_y, a_z] = -\frac{GM}{r^3}\,\left[x, y, z\right]=-\frac{GM}{r^3}\,\vec{r}=-\frac{GM}{r^2}\,\vec{r}_{\rm norm},$$
where $\vec{r}_{\rm norm}=\vec{r}/r$. 

```python
G = 6.67259e-08 # gravitation constant in the cgs units
M = 1.9891e33 # mass of the Sun in grams

def acc_point_mass(x):
    '''Computes acceleration from a point mass at the position with coordinates in x'''
    r = np.sqrt(x[0]**2 + x[1]**2 + x[2]**3)
    return -G*M/r**3 * x
```

## <font color='blue'>A test: idealized circular motion of the Earth around the Sun</font>

### <font color='blue'> Initialize coordinates and velocity components, specify step size, run a sequence of steps</font>

Circular velocity $v_{\rm circ}$ - velocity with which a particle will orbit in a circle around a gravitating mass $M$ at a distance $r$ - can be obtained by equating the centripetal and gravitational forces: 

$$\frac{mv_{\rm circ}^2}{r} = \frac{GMm}{r^2}\ \ \rightarrow\ \ v_{\rm circ} = \sqrt{\frac{GM}{r}}.$$

[Astronomical Unit (AU)](https://en.wikipedia.org/wiki/Astronomical_unit) is approximately equal to the distance between the Earth and the Sun.  

```python
# initial conditions
AU = 1.49598e+13 # 1 AU = average distance from the Earth to the Sun in centimeters (cm)
G = 6.67259e-08 # universal gravitational constant in cgs units 
msun = 1.9891e33 # mass of the Sun in grams
vcirc = (G*msun/AU)**0.5 # circular velocity = sqrt(G*Msun/AU)

x0 = [AU, 0., 0.] # coordinates of the Earth 
v0 = [0., vcirc, 0.] # initial velocity of the Earth 

yr =  365.25 * 24 * 60 * 60 # 1 year in seconds

t0 = 0.
tend = 10 * yr
dt = (tend - t0) / 10000. 

t, xt, vt = time_stepping(x0, v0, t0, tend, dt, acc_func=acc_point_mass)

t /= yr
xt /= AU
```

```python
from matplotlib.animation import FuncAnimation

def animate(i, scat, xp):
    '''
    produce scatter using particle coordinates at step i
    '''
    scat = scat.set_offsets(np.c_[xp[i,:,0], xp[i,:,1]])
    return scat

def pmotion2d(xt=None, pmarkers='.', psizes=10, pcolors='slateblue', 
              xlabel=r'$x\rm\ (AU)$', ylabel=r'$y\rm\ (AU)$',
              interval=30, ssun = 50, lines2plot=None, grid=False, 
              xlims=[-1,1], ylims=[-1,1], figsize=(3,3)): 
    '''
    helper function to show animation of particle motion in 2D using pre-computed
    evolutionary tracks
    '''
    nsteps = xt[:,0].size
    xp = np.zeros((nsteps, 2, 3))
    xp[:,1,0], xp[:,1,1], xp[:,1,2] = xt[:,0], xt[:,1], xt[:,2] 
    # set up figure parameters and axis labels
    fig, ax = plt.subplots(figsize=figsize)

    ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
    ax.set(xlim=xlims, ylim=ylims)
    
    # plot lines and orbits supplied as input, if any
    if lines2plot is not None: 
        for line in lines2plot:
            ax.plot(line[0], line[1], lw=0.5)
    # initialize scatter plot
    ax.scatter(0, 0, s=ssun, marker='o', facecolors='none', edgecolors='orangered') 
    scat = ax.scatter(xp[0,:,0] , xp[0,:,1], s=psizes, c=pcolors, marker=pmarkers) 

    if grid: ax.grid(ls='dotted', lw=0.25)
    anim = FuncAnimation(fig, animate, interval=interval, fargs=(scat, xp),
                         frames=np.shape(xp)[0], blit=False, repeat=True)
    #plt.draw()
    plt.show()
    return anim
```

### <font color='blue'>What's going on?</font>

the 1st order Euler integration step:
$$x_{n+1} = x_n + v_{x,n}\Delta t,$$
$$v_{x,n+1} = v_{x,n} + a_{x,n}\Delta t.$$

while what we do if we switch the order of $x$ and $v$ updates in the code is 

$$\begin{aligned}v_{x,n+1} = &\ v_{x,n} + a_{x,n}\Delta t,\\
x_{n+1}= &\ x_{n} + v_{x,n+1}\Delta t,\end{aligned}$$

which is different. 


This is equivalent to using a more accurate approximation for $x(t)$ using an additional term in the Taylor expansion. 
$$\begin{aligned}x_{n+1}= &\ x_{n} + v_{x,n}\Delta t + \frac{1}{2} a_{x,n}\Delta t^2.\\
v_{x,n+1} = &\ v_{x,n} + a_{x,n}\Delta t.\end{aligned}$$

However, we can notice that the first equation can be written as $x_{n+1} = x_n + v_{x,n+1}\Delta t$ and in the scheme expressed like this computation of $v_{x,n+1}$ is duplicated in both coordinate and velocity update. We can avoid the duplication simply by computing $v_{x,n+1}$ first and then using it to update $x_n$. 



### <font color='blue'>"***Kick-drift (KD)***" integration step.</font>

We can think of this as a scheme with 2 sub-steps within a given step. In the first
substep we keep coordinate the same, but update velocity as though we "kicked'' the particle, but it does not move yet. In the second sub-step, "drifts'' from $x_n$ to $x_{n+1}$ at constant velocity. Such update is called the "***Kick-drift (KD)***" integration step:

1. $x^\ast = x_n;\ \ \ v_{x}^\ast = v_n + a_{x,n}\Delta t$. 
2. $x_{n+1} = x_n + v_{x}^\ast\Delta t;\ \ \ v_{x,n+1} = v_x^\ast$.

### <font color='blue'>"Symplectic" vs non-symplectic update schemes</font>

It turns out that the schemes such as the KD update avoid long-term accumulation of calculation error in positions and velocities (see Section 6.1.17 in the [notes](https://astro.uchicago.edu/~andrey/classes/a211w24/html/nbody/trajint.html)). 

Because total energy of a particle depends on the position (potential energy) and velocity (kinetic energy), this means that the schemes avoid long-term accumulation of errors in the total energy and so confirm to the energy conservation law, which governs dynamics of particles in a gravitational field. Such update schemes are called "symplectic". 

KD is just one "symplectic" scheme. There are others (see Section 6.1 in the [notes](https://astro.uchicago.edu/~andrey/classes/a211w24/html/nbody/trajint.html)).

### <font color='blue'>Force "softening"</font>

Instead of 
$$
\begin{aligned}
a_x =& -\frac{\partial\phi}{\partial x} = \frac{\partial}{\partial x}\left(\frac{GM}{r}\right)\\ =&-\frac{GM}{r^2}\frac{\partial r}{\partial x}=-\frac{GM}{r^3}\, x
\end{aligned}$$
where $r=\sqrt{x^2 +y^2 + z^2},$ ue $r=\sqrt{x^2 +y^2 + z^2 + \epsilon^2}$, where $\epsilon$ is called softening scale - it prevents acceleration from increasing to very large values as $r\rightarrow 0$. 

Force softening is an example of an artifical modification of exact mathematical equations governing the physical system in numerical modeling. It mitigates a problem with force increase and required very small time steps, but tests are needed to ensure that its use does not change behavior of the system qualitative to give us wrong conclusion. Usually, tests are used to determine optimal value of the softening scale and step size for a given problem. 

An additional way to mitigate the issue is to allow step size to adjust and become smaller as acceleration grows. 

### <font color='blue'>Acceleration from spherically symmetric extended mass distribution</font>

The [Navarro-Frenk-White profile](https://en.wikipedia.org/wiki/Navarro%E2%80%93Frenk%E2%80%93White_profile) (NFW) approximately describes density distribution around galaxies. In this approximation mass distribution is *spherically-symmetric* and depends only on the radius $r$ (distance to the center). The density at the distance $r$ is given by 

$$\rho(r) = \frac{\rho_0}{(r/r_s)\left(1+r/r_s\right)},$$

where $\rho_0$ and $r_s$ are characteristic density and scale radius - the parameters of the profile. 

Acceleration due to this mass distribution at the location $\vec{r}$ is 
$$\vec{a} =-\frac{GM_{200}}{\ln(1+c) - c/(1+c)}\,\frac{\frac{r}{r+r_s}-\ln\left(1+\frac{r}{r_s}\right)}{r^3}\, \vec{r}
$$

where mass $M_{200}$ is the mass within the radius $R_{200}$ (also parameters), and $c=R_{200}/r_s$ is "concentration" of the profile. 

```python
msun = 1.9891e33 # mass of the Sun in grams
kpc = 3.08568e+21 # kiloparsec in cm
G = 6.67259e-08 # gravitational constant in the cgs units 

def acc_nfw(x, *args):
    '''Computes acceleration due to the NFW mass distribution at the distance x from the origin
    Parameters:
       x: arrays with 3 coordinates in the system where center of the NFW potential is at the origin
    Returns:
       numpy array with 3 components of the acceleration
    '''
    m200, c = args[0], args[1]
    r200 = 206.285 * (m200/1e12)**(1./3) # R200 in kpc
    rs = r200 / c * kpc
    r = np.linalg.norm(x) 
    acc = G * m200 * msun / (np.log(1+c) - c/(1+c)) * (r/(r+rs) - np.log(1+r/rs))/r**3 * x
    return acc
```

```python
def fnfw(x): 
    '''
    Helper function for computing mass within a radius r for the NFW profile
    '''
    assert(x >= 0)
    return np.log(1.+x) - x/(1+x)

def mw_nfw_ic(r0=8, m200=9e11, c=10):
    '''Compute circular velocity for input distance r0, m200, and concentration c assuming NFW profile'''
    r200 = 206.285 * (m200/1e12)**(1./3) # R200 in kpc
    
    r0 = r0 * kpc
    rs = r200 / c * kpc

    mr0 = m200 * fnfw(r0/rs) / fnfw(c)

    vcirc = (G*mr0*msun / r0)**0.5
    return vcirc
```

```python
# initial conditions
r0 = 8 
m200, c = 9e11, 10 # total mass of the Milky Way in Msun and halo concentration
args = (m200, c) 
vcirc = mw_nfw_ic(r0=r0, m200=m200, c=c)

x0 = [r0*kpc, 0., 0.] # coordinates of the Earth 
v0 = [0., vcirc, 0.] # initial velocity of the Earth 

t0 = 0.
tend = 1e9 * yr # duration of calculation
dt = (tend - t0) / 10000. 


t, xt, vt = time_stepping(x0, v0, t0, tend, dt, *args, acc_func=acc_nfw)

t /= (1e6*yr)
xt /= kpc
```
