## <font color='blue'>The perceptron neuron</font>

Based on nodes that perform some operations on their inputs to produce outputs. Different types of nodes are used in deep learning, the simplest of which are ***perceptron neurons.*** 
<p>
<img width=900 src="https://appliedgo.net/media/perceptron/neuron.png
"></img>


### <font color='blue'>A single perceptron neuron with multiple inputs outputs.</font> 
    
If $\mathbf{x}$ is input and $\mathbf{y}$ is output, the neuron is supposed to approximate a function $\mathbf{y}=f(\mathbf{x})$ with a function $g(\mathbf{x})$ that it can represent:
     
$$
y=f(\mathbf{x})\approx g(\mathbf{x})=\begin{cases} 
1, {\rm if}\ \mathbf{w}\cdot\mathbf{x}  + b>0,\\
0, {\rm otherwise}\end{cases}
$$

Here $\{w_j\}$ are weights, $b$ is the bias, and $\mathbf{x}=\{x_j\}$ are input values from the neurons in the previous layer in the network. 

In general: $\mathbf{w}\cdot\mathbf{x}=\sum\limits_{i=0}^{N_{\rm inputs}}w_i\,x_i$ is the scalar product of vectors $\mathbf{w}$ and $\mathbf{x}$.


```python
def perceptron(x):
    if np.isscalar(x):
        if x > 0:
            return 1.
        else:
            return 0.
    else:
        out = np.zeros_like(x)
        out[x>0] = 1.
        return out
```

```python
xtr = np.linspace(0, 1, 100)
ftr = np.exp(xtr)
```

```python
fapprox = 0.7 + 2.6*perceptron(3.*xtr -2)

plt.figure(figsize=(3,3))
plt.xlim(0., 1)
plt.xlabel('$x$'); plt.ylabel('$\sigma(x)$')
plt.plot(xtr, fapprox)
plt.plot(xtr, ftr, ls='--', c='orangered')
plt.grid(ls='dotted')
```

```python
def mse0(wb, *args):
    wout, bout, w, b = wb
    x, f = args[0], args[1]
    return np.sum((bout + wout*perceptron(x*w + b) - f)**2)
```

```python
from scipy.optimize import minimize

wb0 = np.random.uniform(size=4)
res = minimize(mse0, wb0, method='Powell', args=(xtr, ftr))
```

```python
wout, bout, w, b = res.x
fapprox = bout + wout*perceptron(xtr*w + b)

plt.figure(figsize=(3,3))
plt.xlim(0., 1)
plt.xlabel('$x$'); plt.ylabel('$f(x)$')
plt.plot(xtr, fapprox)
plt.plot(xtr, ftr, ls='--', c='orangered')
plt.grid(ls='dotted')
```

### <font color='blue'>Sigmoid neurons</font>

Sigmoid neuron: $\mathrm{output}=\sigma(\mathbf{w}\cdot \mathbf{x} + b)$, where ***sigmoid function*** $\sigma$ is chosen to approximate a step from $0$ to some value. 


<img width=500 src="https://upload.wikimedia.org/wikipedia/commons/thumb/8/88/Logistic-curve.svg/1920px-Logistic-curve.svg.png"></img>


### <font color='blue'>Types of activation functions often used</font>

The classic sigmoid $\sigma()$ is 

$$\sigma(z) = \frac{1}{1+e^{-z}}.$$
    
    
$$z = \mathbf{w}\cdot\mathbf{x} + b = \sum\limits_{i=0}^{N_{\rm inputs}}w_i\,x_i + b.$$

Other examples are $\mathrm{tanh}$ and ReLU (Rectified Linear Unit). 
    
    
$$\mathrm{ReLU}(z)=\max(0,z)$$
    
or softmax (used in output layers in classification problems):

$$\mathrm{softmax}(z) = \frac{e^{\mathbf{z}}}{\sum_i e^{z_i}}.$$
    

```python
def sigmoid(x):
    return 1.0/(1.0 + np.exp(-x))

def ReLU(x):
    return np.maximum(0.,x)

def softmax(x):
    return np.exp(x)/np.sum(np.exp(x))

def sigmoid_illustration():
    z = np.linspace(-10., 10., 200)

    _, ax = plt.subplots(1,4, figsize=(12,3))
    ax[0].set_xlabel(r'$z$'); 
    ax[0].set_ylabel(r'$\sigma(z)$')
    ax[0].plot(z, sigmoid(z), lw=2.); ax[0].set_title('sigmoid/logistic')
    ax[1].set_xlabel(r'$z$'); 
    ax[1].plot(z, np.tanh(z), lw=2.); ax[1].set_title('tanh')
    ax[2].set_xlabel(r'$z$'); 
    ax[2].plot(z, ReLU(z), lw=2.); ax[2].set_title('ReLU')
    ax[3].set_xlabel(r'$z$'); 
    ax[3].plot(z, softmax(z), lw=2.); ax[3].set_title('softmax')

    plt.show()
```

```python
sigmoid_illustration()
```

```python
def mse1(wb, *args):
    wout, bout, w, b = wb
    x, f = args[0], args[1]
    return np.sum((bout + wout*sigmoid(x*w + b) - f)**2)
```

### <font color='blue'>Can we approximate distance $d_L$ as a function of $\Omega_{\rm m}$ with a single sigmoid neuron?</font>


```python
from astropy.cosmology import LambdaCDM
import astropy.units as u

def d_l_astropy(z, H0, Om0, OmL):
    cosmo = LambdaCDM(H0=H0, Om0=Om0, Ode0=OmL)
    return cosmo.luminosity_distance(z=z) / u.Mpc
```

```python
ntrain, ntest = 10, 1000 # number of the training set and test points
om0min, om0max = 0., 1.
om0train = np.linspace(om0min, om0max, ntrain)
om0test = np.linspace(om0min, om0max, ntest+1)

# evaluate d_L at training points
dltrain = np.empty_like(om0train)
for i, omd in enumerate(om0train):
    dltrain[i] = d_l_astropy(0.1, 70., omd, 1.-omd) 
# compute 
dltest = np.empty_like(om0test)
for i, omd in enumerate(om0test):
    dltest[i] = d_l_astropy(0.1, 70., omd, 1.-omd) 
```

```python
wb0 = np.random.uniform(size=4)
res = minimize(mse1, wb0, method='Powell', args=(om0train, dltrain))

wout, bout, w, b = res.x
dl_approx = bout + wout*sigmoid(om0test*w + b)

```

```python
from codes.plotting import plot_line_points

plot_line_points(om0test, dl_approx, xp=om0train, yp=dltrain, points=True, psize=40, 
                 xlabel=r'$\Omega_{\rm m0}$', ylabel=r'$d_L\rm\ (Mpc)$', 
                 legend='cubic spline', plegend='training values',
                 legendloc='upper right', figsize=3)
```

```python
def fwiggly(x):
    """
    A wiggly function for illustration from Michael Nielsen's 
    "Neural Networks and Deep Learning" book: 
    http://neuralnetworksanddeeplearning.com/chap4.html
    """
    
    return 0.2 + 0.4*x**2 + 0.3*x*np.sin(15.*x) + 0.05*np.cos(50*x) - 0.5
```

```python
ntrain = 20
xtr = np.sort(np.random.uniform(0., 1., ntrain))
xtr = np.linspace(0., 1., 100)
ftr = fwiggly(xtr)

wb0 = np.random.uniform(size=4)
res = minimize(mse1, wb0, method='Powell', args=(xtr, ftr))
```

```python
wout, bout, w, b = res.x
fapprox = bout + wout*sigmoid(xtr*w + b)

plt.figure(figsize=(3,3))
plt.xlim(0., 1)
plt.xlabel('$x$'); plt.ylabel('$w(x)$')
plt.plot(xtr, fapprox)
plt.plot(xtr, ftr, ls='--', c='orangered')
plt.grid(ls='dotted')
```

### <font color='blue'>Let's consider a network with 2 neurons in the hidden layer and a single neuron in the output layer</font>
    
Outputs of the 2 hidden layer neurons given input $x$: 

$$x_1 = \sigma(w_{11} x +b_{11});\ \ \ x_2 = \sigma(w_{12}x + b_{12}).$$

Output of the output layer neuron: 

$$
y=f(\mathbf{x})\approx g(\mathbf{x}) = \sigma(\mathbf{w}\cdot\mathbf{x}+ b)=\sigma\left(w_{21}x_1 + w_{22}x_2 + b_{21}\right),
$$


<p>
<center>
<img width=500 src="https://astro.uchicago.edu/~andrey/classes/a211/img/nn_2neurons.png"></img>

```python
def mse2(wb, *args):
    wout1, bout1, wout2, bout2, w1, b1, w2, b2 = wb
    x, f = args[0], args[1]
    return np.sum((bout1 + wout1*sigmoid(x*w1 + b1) + bout2 + wout2*sigmoid(x*w2+b2) - f)**2)
    
```

```python
wb0 = np.random.uniform(size=8)
res = minimize(mse2, wb0, method='Powell', args=(om0train, dltrain))

wout1, bout1, wout2, bout2, w1, b1, w2, b2 = res.x
dl_approx = bout1 + wout1*sigmoid(om0test*w1 + b1) + bout2 + wout2*sigmoid(om0test*w2+b2)

```

### <font color='blue'>2 inputs = bivariate function, the number of hidden neurons = the number of terms in the expansion</font>


```python
def plot_sigmoid2d(w1=5, w2=-15, b=5.5):
    fig = plt.figure(figsize=(3,3))
    ax = fig.add_subplot(111,projection='3d')
    plt.xlabel('$x_1$'); plt.ylabel('$x_2$'); ax.set_zlabel('$z$')

    x1, x2 = np.meshgrid(np.linspace(0.,1.,200), np.linspace(0.,1.,200))
    z = np.zeros_like(x1)
    z = 1./(1+np.exp(-(w1*x1 + w2*x2 + b)))
    # plot the plane
    ax.plot_surface(x1, x2, z, alpha=0.5)

    plt.show()
```

```python
%matplotlib inline

plot_sigmoid2d(w1=1, w2=-15, b=5.5)
```

### <font color='darkblue'>Using neural network for hand-written digit classification
    
Example using scikit-learn <a href="https://scikit-learn.org/stable/modules/generated/sklearn.neural_network.MLPClassifier.html">MLPClassifier</a> class and its methods. 

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_openml

# Load data from https://www.openml.org/d/554
X, y = fetch_openml('mnist_784', version=1, return_X_y=True)

# normalize images, split data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X/255., y, test_size=0.2, random_state=1)
```

```python
from sklearn.neural_network import MLPClassifier

tstart = default_timer()
mlp = MLPClassifier(max_iter=20, alpha=1.e-4, solver='sgd', random_state=1,
                    learning_rate_init=0.1, verbose=10, 
                    hidden_layer_sizes=(784,100,10)).fit(X_train, y_train)

print('completed in {:.2f} sec'.format(default_timer()-tstart))
```

```python
from sklearn.metrics import accuracy_score
print('Accuracy on the training data is {:.2f}%'.format(100.*accuracy_score(y_train, mlp.predict(X_train))))
print('Accuracy on the test data is {:.2f}%'.format(100.*accuracy_score(y_test, mlp.predict(X_test))))
```

We can see that with a relatively simple network with one hidden later (100 nodes), the neural net reached >98.5% classification accuracy. In fact, this accuracy is better than accuracy of any method we experimented with in this notebook. You can note that accuracy on the training data is 100%, so the neural net learned to classify training data perfectly. However, 1.5% of images in test data is missclassified. 

This gives you a glimpse of why neural networks are popular in machine learning. If one could overcome challenges of their design and training, results they can often provide are better than obtainable with any other methods. Although neural nets are often not competitve in small number of dimensions, they scale better than other methods to very large numbers of dimensions and are now often applied in spaces with tens of millions of dimensions. 

As any ML method, however, neural networks for a given problem work better in smaller number of dimensions. So it's worthwhile to reduce dimensionality of the problem as much as possible, by reducing image size of by pruning function variables that are used in approximation (limiting number of features, or feature selection, in the ML jargon).

```python
from PIL import Image
import requests
from io import BytesIO

def read_image_from_url(url):

    response = requests.get(url)
    img = Image.open(BytesIO(response.content))
    return np.array(img)
```

```python
def plot_images(img1, img2, figsize=3):
    fig, axs = plt.subplots(1, 2, figsize=(2*figsize,figsize))
    for ax, img in zip(axs,[img1,img2]):
        ax.axis("off")
        ax.imshow(img, cmap='gray')
    plt.show()
    
```

```python
import numpy as np
from numpy.lib.stride_tricks import as_strided


def pool2d(A, kernel_size=2, stride=1, padding=0, pool_mode='max'):
    '''
    2D Pooling

    Parameters:
        A: input 2D array
        kernel_size: int, the size of the window over which we take pool
        stride: int, the stride of the window
        padding: int, implicit zero paddings on both sides of the input
        pool_mode: string, 'max' or 'avg'
    '''
    # Padding
    A = np.pad(A, padding, mode='constant')

    # Window view of A
    output_shape = ((A.shape[0] - kernel_size) // stride + 1,
                    (A.shape[1] - kernel_size) // stride + 1)

    shape_w = (output_shape[0], output_shape[1], kernel_size, kernel_size)
    strides_w = (stride*A.strides[0], stride*A.strides[1], A.strides[0], A.strides[1])

    A_w = as_strided(A, shape_w, strides_w)

    # Return the result of pooling
    if pool_mode == 'max':
        return A_w.max(axis=(2, 3))
    elif pool_mode == 'avg':
        return A_w.mean(axis=(2, 3))
```

```python
from io import BytesIO
import requests
from PIL import Image


def read_image_from_url(url):
    response = requests.get(url)
    img = Image.open(BytesIO(response.content))
    return np.array(img)

def plot_images(img1, img2, figsize=3):
    from matplotlib import pylab as plt
    fig, axs = plt.subplots(1, 2, figsize=(2*figsize,figsize))
    for ax, img in zip(axs,[img1,img2]):
        ax.axis("off")
        ax.imshow(img, cmap='gray')
    plt.show()
```

```python
# "gradient" nxn kernel
n = 10
grad = np.ones((n,n))
grad[:,0:n//2] = -1.

# averaging nxn kernel
n = 10
avg = np.ones((n,n))

# averaging 3x3 kernel
avg_cross =  np.array([[0,   1/5,   0],
                       [1/5, 1/5, 1/5],
                       [0,    1/5,  0]])

# random
rnd = np.random.normal(size=(3,3))
```

```python
# horizonthal 3x3 Sobel filter
h_sobel = np.array([[1, 2, 1],
                    [0, 0, 0],
                    [-1, -2, -1]])
# vertical 3x3 Sobel filter
v_sobel = np.array([[1, 0, -1],
                    [2, 0, -2],
                    [1, 0, -1]])
# egde-detector
edge  = np.array([[-1,-1,-1],
                  [-1, 8,-1],
                  [-1,-1,-1]])
```

```python
from scipy.signal import convolve2d

url = 'https://astro.uchicago.edu/~andrey/classes/a205/img/uchicago_logo_gs.jpg'
img = read_image_from_url(url)[:,:]

kernels = ['avg', 'avg_cross', 'grad', 'h_sobel', 'v_sobel', 'edge', 'random']

for kernel, k in zip(kernels, [avg, avg_cross, grad, h_sobel, v_sobel, edge, rnd]):
    print(f'kernel: {kernel:s}')
    imgc = convolve2d(img, k, boundary='symm', mode='same')
    plot_images(img, np.abs(imgc), figsize=2)
```

```python
print(f'original image size: {img.shape}\n')

for pool in ['avg', 'max']:
    img_pooled = pool2d(img, kernel_size=3, stride=3, pool_mode=pool)
    print(f'pool mode: {pool:s}')
    print(f'pooled image size: {img_pooled.shape}')

    plot_images(img, img_pooled, figsize=2)
```

```python
from scipy.signal import convolve2d

img = np.array(X_train)[3].reshape(28,28)
imgc = convolve2d(img, edge, boundary='symm', mode='same')
imgc = convolve2d(imgc, avg, boundary='symm', mode='same')
imgc = convolve2d(imgc, avg, boundary='symm', mode='same')
imgc = pool2d(imgc, kernel_size=2, stride=2, pool_mode='avg')
imgc = pool2d(imgc, kernel_size=2, stride=2, pool_mode='max')
#print(img.shape, imgc.shape)
plot_images(img, np.abs(imgc))
```

In practice, each convolutional layer is usually followed by [pooling layer](https://computersciencewiki.org/index.php/Max-pooling_/_Pooling) which downsamples or averages the output of the convolutional layer to suppress noise or small variations due to motion of objects in image or other small insignificant variations.

### <font color='blue'>A typical structure of a CNN</font>

A sequence of the convolutional layers (each consisting of multiple kernels) and pooling layers that create abstracted representations of an input image. The output of the last conv/pool layer is passed onto a standard "fully-connected" (FC) layers of sigmoid-like neurons and their output is transformed into a suitable form by the output layer. 

<p>
<center>
<img src="https://miro.medium.com/v2/resize:fit:2000/format:webp/1*Vior6yy6_qN5rBv5Os0kGA.png"></a>

### <font color='blue'>Using values in 3 pixels as coordinates in 3D space</font>

Different "digits" are located in different regions of this space. Thus the job of a neural network is to approximate the digit class in different parts of this space, so that the digit can be predicted accurately for new inputs.   

<center>
<img width=600 src="https://astro.uchicago.edu/~andrey/classes/a211w24/html/_images/pix3_digits_3d.svg"></img>
</center>

### <font color='blue'>Convolution/pooling operations can abstract actual digits into new few-pixel but still distinct "digits"</font>

The boundary in the lower-dimensional space of such low-res digits can then be approximated using a dense layer of sigmoid neurons. 

[An interactive illustration](https://adamharley.com/nn_vis/cnn/2d.html) of the hand-written digit classification with the CNN network. 

### <font color='blue'>Convolution neural networks can be though of having two stages</font>

1. Conv/pool layers that "learn" to represent a low-resolution abstract version of the input images in a way that maximizes accuracy of classification. 

2. Dense layer(s) that approximate distribution of different classes of input images in this lower dimensional space.

## Load a dataset

Load and prepare the [MNIST dataset](http://yann.lecun.com/exdb/mnist/). Convert the sample data from integers to floating-point numbers:

```python
from keras.datasets import mnist

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

train_images = train_images.reshape((60000, 28, 28, 1))
train_images = train_images.astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1))
test_images = test_images.astype('float32') / 255

print(train_labels[0])
# create categorial labels of what they images are to the "one-hot" encoding" categorical labels
train_labels = tf.keras.utils.to_categorical(train_labels)
print(train_labels[0]) 
test_labels = tf.keras.utils.to_categorical(test_labels)
```

```python
from keras import layers
from keras import models

model = models.Sequential() # define that model is Sequential, not Recurrent
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1))) # 2D convolutional layer
model.add(layers.MaxPooling2D((2, 2))) # MaxPooling layer
model.add(layers.Conv2D(64, (3, 3), activation='relu')) # convolutional layer
model.add(layers.MaxPooling2D((2, 2))) # MaxPooling layer
model.add(layers.Conv2D(64, (3, 3), activation='relu')) # convolutional later
model.add(layers.Flatten()) 
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))
```

```python
t0 = default_timer()
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', 
              metrics=['accuracy'])
model.fit(train_images, train_labels, epochs=10, batch_size=64)
    
print(f'with CPU trained in {default_timer()-t0:.3f} sec')
```

```python
model.evaluate(test_images, test_labels)
```

The same using GPU, which requires tensorflow with GPU support enabled and a laptop/computer with GPU present. 

```python
from keras import layers
from keras import models

model = models.Sequential() # define that model is Sequential, not Recurrent
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1))) # 2D convolutional layer
model.add(layers.MaxPooling2D((2, 2))) # MaxPooling layer
model.add(layers.Conv2D(64, (3, 3), activation='relu')) # convolutional layer
model.add(layers.MaxPooling2D((2, 2))) # MaxPooling layer
model.add(layers.Conv2D(64, (3, 3), activation='relu')) # convolutional later
model.add(layers.Flatten()) 
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))
```

The code below will fail if you do not have GPU on your laptop and Tensorflow with GPU support installed. Skip two cells below to the run with CPU if you want to run it. 

```python
device_name = "gpu"
if device_name == "gpu":
    device_name = "/gpu:0"
else:
    device_name = "/cpu:0"

t0 = default_timer()
with tf.device(device_name):
    model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])
    model.fit(train_images, train_labels, epochs=15, batch_size=64)
    
print(f'with {device_name:s} trained in {default_timer()-t0:.3f} sec')
```

```python
model.evaluate(test_images, test_labels)
```

With CNNs we are able to reach 99.20\% classification accuracy after only 15 epochs of training.
