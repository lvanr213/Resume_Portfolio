### <font color='blue'>$N$-body problem</font>

Calculating trajectories of $N\geq 2$ particles (bodies) as was done previously in an external gravitational field, but computing acceleration not from the specified potential of some mass distribution (or at least not just such accelerations), but also acceleration due to gravitational forces particles exert
on each other. 

<p>
<center>
</img>
</center>


The only thing that changes is that now we may have to deal with many particles, so writing code the way we did before will not work and accelerations have to computed by summation of forces from all other particles via a loop. 



### <font color='blue'>Acceleration for any particle $i$ in an $N$-body system:</font>


$$\vec{a}_i = -G\,\sum\limits_{j=1, j\ne i}^N\frac{m_j}{r^3_{ji}}\,\vec{r}_{ji}.$$

Separately for $x$, $y$, and $z$ components of $\vec{a}$: 

$$a_{x,i}=G\sum\limits_{j=0,j\ne i}^{N-1} \frac{m_j}{r^3_{ij}}\,(x_{j}-x_{i}),$$

$$a_{y,i}=G\sum\limits_{j=0,j\ne i}^{N-1} \frac{m_j}{r^3_{ij}}\,(y_{j}-y_{i}),$$

$$a_{z,i}=G\sum\limits_{j=0,j\ne i}^{N-1} \frac{m_j}{r^3_{ij}}\,(z_{j}-z_{i}),$$

$\vec{r}_{ij}$ is vector pointing from particle $j$ to particle $i$ and length equal to the distance between the particles: 

$$r_{ij} = \left[(x_i-x_j)^2 + (y_i-y_j)^2 + (z_i-z_j)^2\right]^{1/2}$$


### <font color='blue'>To implement a code to evolve an $N$-body system:</font>

We can use the time stepping code used to evolve particle in an external gravitation field and 

1. Think about and implement data structures for particles masses, coordinate and velocity components that can handle $N$ particles instead of one. 

2. Implement and use a new separate function that computes and returns acceleration for all particles and pass this function to the time stepping function. 



### <font color='blue'>$N=2$ can be used for testing</font>

Acceleration components for particle 1 (similar for particle 2): 

$$\vec{a}_1 = -G\,\frac{m_2}{r^3_{12}}\,\vec{r}_{21}=G\,\frac{m_2}{r^3_{12}}\,\vec{r}_{12},$$

or 
$$a_{x,1}=G\,\frac{m_2}{r^3_{12}}\,(x_{2}-x_{1}),$$

$$a_{y,1}=G\,\frac{m_2}{r^3_{12}}\,(y_{2}-y_{1}),$$

$$a_{z,1}=G\,\frac{m_2}{r^3_{12}}\,(z_{2}-z_{1}),$$
    
and where $r_{12}$ is the distance between the two particles: 
    
$$r_{12}=\left[(x_1-x_2)^2 + (y_1-y_2)^2 +(z_1-z_2)^2 \right]^{1/2}.$$
  

Equations for particle 2 are similar and can be obtained by switching indices 1 and 2 in the equations above. 

We start with the case of 2 particles both because it is simpler and because for this case, we know solution for orbits from a given initial condition exactly, which allows us to check our computations. We will then proceed to a general case of $N$ particles (see below). 

### <font color='blue'>Some relevant constants and quantities</font>

```python
AU = 1.49598e+13 # 1 AU = average distance from the Earth to the Sun in centimeters (cm)
G = 6.67259e-08 # universal gravitational constant in cgs units 
yr =  3.15569e+07 # 1 year in seconds
msun = 1.9891e33 # mass of the Sun in grams
mearth = 5.9742e27 # mass of the Earth in grams
vcirc = (G*msun/AU)**0.5 # circular velocity = sqrt(G*Msun/AU)

```

```python
def acc_2body(x1, y1, z1, x2, y2, z2):
    r12 = ((x1-x2)**2 + (y1-y2)**2 + (z1-z2)**2)**0.5
    ax1 = G * m2 / r12**3 * (x2 - x1)
    ay1 = G * m2 / r12**3 * (y2 - y1)
    az1 = G * m2 / r12**3 * (z2 - z1)
    ax2 = G * m1 / r12**3 * (x1 - x2)
    ay2 = G * m1 / r12**3 * (y1 - y2)
    az2 = G * m1 / r12**3 * (z1 - z2)
    return ax1, ay1, az1, ax2, ay2, az2

```

```python
def orbit_integrate_2body(dt, nsteps, r1, r2, v1, v2, m1, m2, *args):
    '''
    Integrate trajectories of 2 gravitationally interacting particles 
    
    Parameters:
        dt - float, time step in seconds
        nsteps - int, the number of steps of size dt to make
        r1 - list with 3 float elements containing coordinates x, y, z of particle 1 in cm
        r2 - list with 3 float elements containing coordinates x, y, z of particle 2 in cm
        v1 - list with 3 float elements containing velocity components vx, vy, vz of particle 1 in cm/s
        v2 - list with 3 float elements containing velocity components vx, vy, vz of particle 2 in cm/s
        
    Returns:
        x1, y1, z1, x2, y2, z2 - NumPy emptys of floats of nsteps size each containing a sequence of coordinates 
            representing trajectory of particles 1 and 2
    '''
    x1, y1, z1 = r1
    x2, y2, z2 = r2
    vx1, vy1, vz1 = v1
    vx2, vy2, vz2 = v2

    # emptys to record coordinates and velocities along the trajectory
    x1t, y1t, z1t = np.empty(nsteps), np.empty(nsteps), np.empty(nsteps)  
    x2t, y2t, z2t = np.empty(nsteps), np.empty(nsteps), np.empty(nsteps)
    vx1t, vy1t, vz1t = np.empty(nsteps), np.empty(nsteps), np.empty(nsteps)  
    vx2t, vy2t, vz2t = np.empty(nsteps), np.empty(nsteps), np.empty(nsteps)
    x1t[0], y1t[0], z1t[0] = x1, y1, z1 
    x2t[0], y2t[0], z2t[0] = x2, y2, z2 
    vx1t[0], vy1t[0], vz1t[0] = vx1, vy1, vz1 
    vx2t[0], vy2t[0], vz2t[0] = vx2, vy2, vz2 

    for n in range(nsteps-1):
        # KICK (update of velocity components)
        ax1, ay1, az1, ax2, ay2, az2 = acc_2body(x1, y1, z1, x2, y2, z2, *args)
        vx1 = vx1 + ax1 * dt 
        vy1 = vy1 + ay1 * dt 
        vz1 = vz1 + az1 * dt 
        vx2 = vx2 + ax2 * dt 
        vy2 = vy2 + ay2 * dt 
        vz2 = vz2 + az2 * dt 
        # DRIFT (coasting update to particle positions)
        x1  = x1 + vx1 * dt 
        y1  = y1 + vy1 * dt 
        z1  = z1 + vz1 * dt 
        x2  = x2 + vx2 * dt 
        y2  = y2 + vy2 * dt 
        z2  = z2 + vz2 * dt
        x1t[n+1], y1t[n+1], z1t[n+1] = x1, y1, z1 
        x2t[n+1], y2t[n+1], z2t[n+1] = x2, y2, z2 
        vx1t[n+1], vy1t[n+1], vz1t[n+1] = vx1, vy1, vz1 
        vx2t[n+1], vy2t[n+1], vz2t[n+1] = vx2, vy2, vz2 
    
    return x1t, y1t, z1t, x2t, y2t, z2t, vx1t, vy1t, vz1t, vx2t, vy2t, vz2t
```

```python
from matplotlib.animation import FuncAnimation

def animate(i, scat, xp):
    '''
    produce scatter using particle coordinates at step i
    '''
    scat = scat.set_offsets(np.c_[xp[i,:,0], xp[i,:,1]])
    return scat

def pmotion2d(xp=None, pmarkers='.', psizes=10, pcolors='slateblue', 
              xlabel=r'$x\rm\ (AU)$', ylabel=r'$y\rm\ (AU)$',
              interval=30, ssun = 50, lines2plot=None, grid=False, 
              xlims=[-1,1], ylims=[-1,1], figsize=(3,3)): 
    '''
    helper function to show animation of particle motion in 2D using pre-computed
    evolutionary tracks
    '''
    # set up figure parameters and axis labels
    fig, ax = plt.subplots(figsize=figsize)

    ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
    ax.set(xlim=xlims, ylim=ylims)
    
    # plot lines and orbits supplied as input, if any
    if lines2plot is not None: 
        for line in lines2plot:
            ax.plot(line[0], line[1], lw=0.5)
    # initialize scatter plot
    ax.scatter(0, 0, s=ssun, marker='o', facecolors='none', edgecolors='orangered') 
    scat = ax.scatter(xp[0,:,0] , xp[0,:,1], s=psizes, c=pcolors, marker=pmarkers) 

    if grid: ax.grid(ls='dotted', lw=0.25)
    anim = FuncAnimation(fig, animate, interval=interval, fargs=(scat, xp),
                         frames=np.shape(xp)[0], blit=False, repeat=True)
    #plt.draw()
    plt.show()
    return anim
```

### <font color='blue'> Initialize positions and velocities for a test: circular motion of the Earth around the Sun</font>

```python
x1, y1, z1 = 0., 0., 0. # coordinates of the Sun
x2, y2, z2 =  AU, 0., 0. # coordinates of the Earth 
r1, r2 = [x1, y1, z1], [x2, y2, z2]

vx1, vy1, vz1 = 0., 0., 0. # initial velocity of the Sun
vx2, vy2, vz2 = 0., vcirc, 0. # initial velocity of the Earth 
v1, v2 = [vx1, vy1, vz1], [vx2, vy2, vz2]
m1, m2 = msun, mearth # masses 

nsteps = 10000 # number of steps
dt =  10 * yr / nsteps # time step

x1, y1, z1, x2, y2, z2, vx1, vy1, vz1, vx2, vy2, vz2 = orbit_integrate_2body(dt, nsteps, r1, r2, v1, v2, m1, m2)
x1, y1, z1, x2, y2, z2 = x1/AU, y1/AU, z1/AU, x2/AU, y2/AU, z2/AU

```

```python
%matplotlib notebook
theta = np.linspace(0., 2.*np.pi, 200)
# coordinates of x y lying on a circle; will be used to plot circle as a line
xl, yl = np.cos(theta), np.sin(theta) 
rmax = 2 # [-rmax, rmax] woill be the range of x and y to plot

# in order to use the function below, particle coordinates should be maked into array xp as follows
xp = np.zeros((nsteps, 2, 3))
xp[:,0,0], xp[:,0,1], xp[:,0,2] = x1, y1, z1
xp[:,1,0], xp[:,1,1], xp[:,1,2] = x2, y2, z2
```

```python
anim = pmotion2d(xp=xp, lines2plot=[[xl,yl]], interval=1, psizes=30, 
                 xlims=[-rmax, rmax], ylims=[-rmax,rmax], grid=True, figsize=(4,4))

```

### <font color='blue'> Plot orbit of the "Earth" and velocity of the "Sun"</font>

```python

fig, ax = plt.subplots(1, 2, figsize=(7,3)) # define sub-panels
fig.subplots_adjust(wspace=0.4)

v1 = (vx1**2 + vy1**2 + vz1**2)**0.5
ax[0].plot(x2, y2)
ax[0].set_ylabel('Y (AU)') # add labels to axes
ax[0].set_xlabel('X (AU)')
ax[1].plot(np.arange(nsteps)*dt/yr, vx1)
ax[1].set_xlabel('t (years)')
ax[1].set_ylabel('v (cm/s)')
plt.show()
```

### <font color='blue'> Initialize positions and velocities for a test: motion of the binary system of Sirius A and B stars</font>

From the reading for the $N=2$ problem: 

$$v_1 = \frac{m_2}{m_1+m_2} v_{\rm per}, $$
$$v_2 = \frac{m_1}{m_1+m_2} v_{\rm per}, $$

For elliptical orbits $r_{\rm per}=a(1-e)$ and 
$$v^2_{\rm per} = G\,\frac{(m_1+m_2)}{a}\,\left(\frac{1+e}{1-e}\right),$$



<p>
<center>
</img>
</center>


This example is from Section 4.3 of the ["Numerical Python in Astronomy and Astrophysics"](https://link.springer.com/book/10.1007/978-3-030-70347-9) book. 

```python
a = 2.64*7.4957*AU # semi-major axis
e = 0.5914 # ellipticity
d_p = a * (1 - e)
v_p = (G*(m1 + m2) / a * (1+e)/(1-e))**0.5
m1, m2 = 2.06*msun, 1.02*msun # masses of Sirius A and B

x1, y1, z1 = -m2/(m1+m2)*d_p, 0., 0. # coordinates of the Sirius A
x2, y2, z2 =  m1/(m1+m2)*d_p, 0., 0. # coordinates of the Sirius B
r1, r2 = [x1, y1, z1], [x2, y2, z2]

vx1, vy1, vz1 = 0.,  -m2/(m1+m2)*v_p, 0. # initial velocity of the Sun
vx2, vy2, vz2 = 0., m1/(m1+m2)*v_p, 0. # initial velocity of the Earth 

v1, v2 = [vx1, vy1, vz1], [vx2, vy2, vz2]

```

```python
nsteps = 1000 # number of steps
dt =  100 * yr / nsteps # time step

x1, y1, z1, x2, y2, z2, vx1, vy1, vz1, vx2, vy2, vz2 = orbit_integrate_2body(dt, nsteps, r1, r2, v1, v2, m1, m2)
x1, y1, z1, x2, y2, z2 = x1/AU, y1/AU, z1/AU, x2/AU, y2/AU, z2/AU
```

```python
%matplotlib notebook
theta = np.linspace(0., 2.*np.pi, 200)
# coordinates of x y lying on a circle; will be used to plot circle as a line
xl, yl = np.cos(theta), np.sin(theta) 
rmax = 25 # [-rmax, rmax] will be the range of x and y to plot

# in order to use the function below, particle coordinates should be maked into array xp as follows
xp = np.zeros((nsteps, 2, 3))
xp[:,0,0], xp[:,0,1], xp[:,0,2] = x1, y1, z1
xp[:,1,0], xp[:,1,1], xp[:,1,2] = x2, y2, z2
```

```python
anim = pmotion2d(xp=xp, lines2plot=[[xl,yl]], interval=10, psizes=30, 
                 xlims=[-rmax, rmax], ylims=[-rmax,rmax], grid=True, figsize=(4,4))
```

```python
%matplotlib inline

fig, ax = plt.subplots(1, 2, figsize=(7,3)) # define sub-panels
fig.subplots_adjust(wspace=0.4)

v1 = (vx1**2 + vy1**2 + vz1**2)**0.5
ax[0].plot(x2, y2)
ax[0].set_ylabel('Y (AU)') # add labels to axes
ax[0].set_xlabel('X (AU)')
ax[1].plot(np.arange(nsteps)*dt/yr, vx1)
ax[1].set_xlabel('t (years)')
ax[1].set_ylabel('v (cm/s)')
plt.show()
```

Let's think what we need to modify in this time-stepping code to allow calculations for $N$ particles. 

```python
def time_stepping(x0, v0, t0, tend, dt, *args, acc_func=None, scheme='kd'):
    # initialize time and arrays to keep coordinate and velocity components
    t = t0
    x, v = np.array(x0), np.array(v0)

    # Initialize lists to store times, coordimates, and velocities along trajectory
    time, xt, vt = [np.copy(t)], [np.copy(x0)], [np.copy(v0)]

    # Time stepping loop
    while t < tend:
        # Calculate acceleration components using the provided function
        a = np.array(acc_func(x, *args))

        # update v and x using KD scheme 
        v += a * dt
        x += v * dt
        
        # Update time
        t += dt

        # Append current state to the trajectory
        time.append(t)
        xt.append(np.copy(x))
        vt.append(np.copy(v))
        
    return np.array(time), np.array(xt), np.array(vt)

```
