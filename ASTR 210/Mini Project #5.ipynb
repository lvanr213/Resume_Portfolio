```python
import numpy as np
import matplotlib.pyplot as plt
theta = np.arange(0.0,2*np.pi, 0.01)
rf = (1 - 0.5* (np.sin(theta * 0.5))**2 ) * (np.cos(theta* 4)) **2
x = rf * np.cos(theta)
y = rf * np.sin(theta)
plt.plot(x,y)
plt.show()
n = 10 
r = []
for n in [10,100,1000,10000,100000,1000000,10000000]:
    x1 = np.random.uniform(-1,1,n)
    y1 = np.random.uniform(-1,1,n)
    theta = np.arctan2(y1,x1)
    r1 = np.sqrt( (x1**2) + (y1**2) )
    rf1 = (1 - 0.5* (np.sin(theta * 0.5))**2 ) * (np.cos(theta* 4)) **2
    a = (np.sum(r1<rf1)/n)
    r.append(a)
print(r)
```
