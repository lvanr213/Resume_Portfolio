```python
import numpy as np
import math
import scipy 
data_filename = "LbolMbh.dat"
Lbol, BHM = np.loadtxt(data_filename, unpack=True, skiprows=1)
print(Lbol)
print(BHM)
def Mbol_func(Lum):
    M = 4.28 - 2.5 * np.log10(Lum / (3.86e33))
    return M

Mbol_lst=[]

for x in Lbol:
    Mbol = Mbol_func(x)
    Mbol_lst.append(Mbol)


print(scipy.stats.pearsonr(Lbol,BHM))
print(scipy.stats.pearsonr(Mbol_lst,BHM))
print(scipy.stats.pearsonr(Mbol_lst,np.log10(BHM)))
print(scipy.stats.pearsonr(Mbol_lst,1/BHM))

print(scipy.stats.spearmanr(Lbol,BHM))
print(scipy.stats.spearmanr(Mbol_lst,BHM))
print(scipy.stats.spearmanr(Mbol_lst,np.log10(BHM)))
print(scipy.stats.spearmanr(Mbol_lst,1/BHM))
```
