```python
%matplotlib notebook
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
```

https://ui.adsabs.harvard.edu/abs/2023A%26A...676A..71M/abstract


```python
d = np.loadtxt("LbolMbh.dat",unpack=True)
Lbol = d[0]/3.846e33
Mbh = d[1]

#fig, ax = plt.subplots()
#ax.plot(t, s)

plt.gca().set_xlabel(r"$L_{\rm BOL} [{\rm L_\odot}]$")
plt.gca().set_ylabel(r"$M_{\rm BH} [{\rm M_\odot}]$")
plt.loglog(Lbol,Mbh,".")
```

```python
Mbol = -2.5*np.log(Lbol) + 4.75

print(scipy.stats.pearsonr(Lbol,Mbh))
print(scipy.stats.pearsonr(Mbol,Mbh))
print(scipy.stats.pearsonr(Mbol,np.log10(Mbh)))
print(scipy.stats.pearsonr(Mbol,1/Mbh))
```

```python
print(scipy.stats.spearmanr(Lbol,Mbh))
print(scipy.stats.spearmanr(Mbol,Mbh))
print(scipy.stats.spearmanr(Mbol,np.log10(Mbh)))
print(scipy.stats.spearmanr(Mbol,1/Mbh))
```
