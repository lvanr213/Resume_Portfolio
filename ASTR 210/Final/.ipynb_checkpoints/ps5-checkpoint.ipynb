```python
%matplotlib notebook
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
```

### 1. (15 points)

```python
# Part A
#

d = np.loadtxt("LbolMbh2.dat",unpack=True)
y = d[0]
x = d[1]

res1 = scipy.stats.linregress(x,y)
print(res1)
```

```python
# Part B
#

xbar = np.mean(x)
ybar = np.mean(y)

a = (np.mean(x*y)-xbar*ybar)/(np.mean(x*x)-xbar**2)
b = ybar - a*xbar
print(a,b)
```

```python
# Part C
#

res2 = scipy.stats.linregress(y,x)
print(res2)
```

```python
# Part D
#

fig, ax = plt.subplots()

ax.set_xlabel(r"$\log10(M_{\rm BH}) [{\rm M_\odot}]$")
ax.set_ylabel(r"$\log10(L_{\rm BOL}) [{\rm L_\odot}]$")

x1 = np.sort(x)
y1 = res1.slope*x1 + res1.intercept

y2 = np.sort(y)
x2 = res2.slope*y2 + res2.intercept

ax.plot(x,y,"k.")
ax.plot(x1,y1,color="r")
ax.plot(x2,y2,color="b")
```

### 2. (15 points)

```python
# Part A
#
def fun1(t):
    '''
    Cost function for y(x) with no errors
    '''
    return (y-t[0]*x-t[1])
#

def fun2(t):
    '''
    Cost function for x(y) with no errors
    '''
    return (x-t[0]*y-t[1])
#

print(scipy.optimize.least_squares(fun1,[0,0]).x)
print(scipy.optimize.least_squares(fun2,[0,0]).x)
```

```python
# Part B
#

d = np.loadtxt("LbolMbh2.dat",unpack=True)
y = d[0]
x = d[1]
sy = d[2]
sx = d[3]

fig, ax = plt.subplots()

ax.set_xlabel(r"$\log10(M_{\rm BH}) [{\rm M_\odot}]$")
ax.set_ylabel(r"$\log10(L_{\rm BOL}) [{\rm L_\odot}]$")

ax.errorbar(x,y,sy,sx,"k.",alpha=0.5)
```

```python
# Part C
#

def lfun1(t):
    '''
    Cost function for y(x) with errors accounted for
    '''
    return (y-t[0]*x-t[1])/sy
#

def lfun2(t):
    '''
    Cost function for x(y) with errors accounted for
    '''
    return (x-t[0]*y-t[1])/sx
#

t1 = scipy.optimize.least_squares(lfun1,[0,0]).x
t2 = scipy.optimize.least_squares(lfun2,[0,0]).x
print(t1)
print(t2)
```

```python
# Part D
#

y1 = t1[0]*x1 + t1[1]
x2 = t2[0]*y2 + t2[1]

ax.plot(x,y,"k.")
ax.plot(x1,y1,color="r")
ax.plot(x2,y2,color="b")
```

```python
# Part E
#

chisq1 = np.sum(lfun1(t1)**2)
chisq2 = np.sum(lfun2(t2)**2)

ndf = len(x) - 2

print(chisq1,1-scipy.stats.chi2.cdf(chisq1,ndf))
print(chisq2,1-scipy.stats.chi2.cdf(chisq2,ndf))
```

### 3. (20 points)

```python
# Part A
#

d = np.loadtxt("hlsp_ps1cosmo_panstarrs_gpc1_all_model_v1_lcparam-full.txt",usecols=[1,4,5],unpack=True,dtype=np.float32)

zmax = 10.03
sel = (d[0] < zmax)
z = np.compress(sel,d[0])
m = np.compress(sel,d[1])
sm = np.compress(sel,d[2])

x = np.log10(z)
y = 0.2*(m+19.269) - 5 # -5 is to convert from 10pc to Mpc
sy = 0.2*sm

fig, ax = plt.subplots()

ax.set_xlabel(r"$\log10(z)$")
ax.set_ylabel(r"$\log10(d/{\rm Mpc})$")

ax.errorbar(x,y,sy,fmt=".",color="gray",alpha=0.2,zorder=-1)
```

```python
# Part B
#

def lfun1(t):
    '''
    Cost function for y(x) with 2 parameters (linear fit)
    '''
    return (y-t[0]*x-t[1])/sy
#

t1 = scipy.optimize.least_squares(lfun1,[0,0]).x
chisq1 = np.sum(lfun1(t1)**2)
print(t1,chisq1,1-scipy.stats.chi2.cdf(chisq1,len(x)-2))
print("H0=%g"%(3e5*10**(-t1[1])))
```

```python
# Part C
#

x1 = np.sort(x)
y1 = t1[0]*x1 + t1[1]

ax.plot(x1,y1,"b-")
```

```python
# Part D
#

def lfun2(t):
    '''
    Cost function for y(x) with 4 parameters (including correction for the expansion)
    '''
    return (y-t[0]*x-t[1]-np.log10(1+t[2]*z+t[3]*z**2))/sy
#

t2 = scipy.optimize.least_squares(lfun2,[0,0,0,0]).x
chisq2 = np.sum(lfun2(t2)**2)
print(t2,chisq2,1-scipy.stats.chi2.cdf(chisq2,len(x)-4))
print("H0=%g"%(3e5*10**(-t2[1])))
```

```python
y1 = t2[0]*x1 + t2[1] + np.log10(1+t2[2]*np.sort(z)+t2[3]*np.sort(z)**2)

ax.plot(x1,y1,"r-")
```

```python
# Part E
#

tv = np.copy(t2)

xvals = np.zeros((101,101))
yvals = np.zeros((101,101))
pvals = np.zeros((101,101))

for i1 in range(101):
    tv[1] = t2[1] + 0.0005*(i1-50)
    for i0 in range(101):
        tv[0] = t2[0] + 0.0005*(i0-50)
        xvals[i0,i1] = tv[0]
        yvals[i0,i1] = 3e5*10**(-tv[1])
        dchisq = np.sum(lfun2(tv)**2) - chisq2
        pvals[i0,i1] = 1 - scipy.stats.chi2.cdf(dchisq,2)
    #
#

fig, ax = plt.subplots()

ax.set_xlabel(r"$a$")
ax.set_ylabel(r"$H_0 [{\rm km/s/Mpc}]$")

ax.contour(xvals,yvals,pvals,levels=[0.05,0.32],colors="b")
```

```python
# Part F
#

tv = np.copy(t2)

def lfun3(t):
    '''
    Cost function for y(x) for marginalization over nuisance parameters
    tv: fixed values for a and b
    t: input, nuisance parameters C and D.
    '''
    return (y-tv[0]*x-tv[1]-np.log10(np.abs(1+t[0]*z+t[1]*z**2)))/sy
#

for i1 in range(101):
    tv[1] = t2[1] + 0.0005*(i1-50)
    for i0 in range(101):
        tv[0] = t2[0] + 0.0005*(i0-50)
        t3 = scipy.optimize.least_squares(lfun3,tv[2:]).x
        dchisq = np.sum(lfun3(t3)**2) - chisq2
        pvals[i0,i1] = 1 - scipy.stats.chi2.cdf(dchisq,2)
    #
#

ax.contour(xvals,yvals,pvals,levels=[0.05,0.32],colors="r")
```

The contours are now way larger. Since we now find the best-fit values for $C$ and $D$ for each pair $(a,b)$, the fit is more likely and hence a larger set of $(a,b)$ now fall inside our contours. Thus, the first way we computed contours was actually cheating - we deliberately reduced the significance away from the best fit and that made our errors artificially smaller.

```python
# Part G
#

#
# Setting a to 1
#
def lfun4(t):
    '''
    Cost function for y(x) for a=1 case
    '''
    return (y-x-t[0]-np.log10(np.abs(1+t[1]*z+t[2]*z**2)))/sy
#

t4 = scipy.optimize.least_squares(lfun4,[0,0,0]).x
chisq4 = np.sum(lfun4(t4)**2)
print(t4,1-scipy.stats.chi2.cdf(chisq4,len(z)-3))

tv = np.array([1,t4[0],t4[1],t4[2]])

pH0 = np.zeros((101,))
vH0 = np.zeros((101,))

for i1 in range(101):
    tv[1] = t4[0] + 0.0003*(i1-50)
    t3 = scipy.optimize.least_squares(lfun3,tv[2:]).x
    dchisq = np.sum(lfun3(t3)**2) - chisq4
    pH0[i1] = 1 - scipy.stats.chi2.cdf(dchisq,1)
    vH0[i1] = 3e5*10**(-tv[1])
#

fig, ax = plt.subplots()

ax.set_xlabel(r"$H_0 [{\rm km/s/Mpc}]$")
ax.set_ylabel(r"$p(H_0)$")

ax.set_xlim(70,76)

ax.plot(vH0,pH0,color="b")
```

```python
#
#  Report the final error
#
idx = np.argmax(pH0)
H0 = vH0[idx]

idx = np.argwhere(np.diff(np.sign(pH0-0.32))).flatten()
print("1-sigma errors: H0=%4.1f +%4.1f -%4.1f"%(H0,vH0[idx[0]]-H0,H0-vH0[idx[1]]))
                                                            
idx = np.argwhere(np.diff(np.sign(pH0-0.05))).flatten()
print("2-sigma errors: H0=%4.1f +%4.1f -%4.1f"%(H0,vH0[idx[0]]-H0,H0-vH0[idx[1]]))

idx = np.argwhere(np.diff(np.sign(pH0-0.003))).flatten()
print("3-sigma errors: H0=%4.1f +%4.1f -%4.1f"%(H0,vH0[idx[0]]-H0,H0-vH0[idx[1]]))
```
