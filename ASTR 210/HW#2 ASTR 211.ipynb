```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import colors
from matplotlib.ticker import PercentFormatter

# Part 1
def xbar(n):
    rng = np.random.default_rng()
    lst = []
    sum = 0
    for i in range(0,n):
        x = rng.random()
        lst.append(x)
        sum = sum + lst[i]
    sample_mean = sum/n
    return sample_mean

# Part 2
# Sample size are the value n = 10, 40, and 160 asked for in the hw
# As stated it calls the xbar funciton 10,000 times 
# producing a unique sample mean each time and appending it to a list

def sample_mean_lst(sample_size):
    mean_lst = []
    for i in range(0,10000):
        x = xbar(sample_size)
        mean_lst.append(x)
    return mean_lst

## Part 3

N_points = 10000
n_bins = 100

# Distrubtion for histograms
dist1 = sample_mean_lst(10)
dist2 = sample_mean_lst(40)
dist3 = sample_mean_lst(160)
# plotting the 3 histograms separately 
fig, axs = plt.subplots(2,2)
axs[0, 0].hist(dist1, bins = n_bins, density = True )
axs[0, 1].hist(dist2, bins = n_bins,density = True )
axs[1, 0].hist(dist3, bins = n_bins,density =True )

# Part 4

# Function that calculates the standard deviation with input n
def standard_dev(n):
    sd = 1 / (np.sqrt(12 * n))
    return sd

x = np.linspace(0,1,200)
# function that computes the normal distribution 
def normal_dist(x,n):
    sd = standard_dev(n)
    return 1/(sd * np.sqrt(2 * np.pi)) * np.exp(-(x - 0.5) ** 2 / (2 * sd ** 2))
 
# Corresponding normal distributions given n 
pdf1 = normal_dist(x,10)
pdf2 = normal_dist(x,40)
pdf3 = normal_dist(x,160)

# Plotting the  Normal Distribution Results
axs[0,0].plot(x,pdf1 , color = 'red')
axs[0,1].plot(x,pdf2 , color = 'blue')
axs[1,0].plot(x,pdf3 , color = 'yellow')

# Setting Titles
axs[0,0].set_title("For n = 10")
axs[0,1].set_title("For n = 40")
axs[1,0].set_title('For n = 160')
```
